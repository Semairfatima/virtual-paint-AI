```python

import cv2
import mediapipe as mp
import numpy as np
import math

# MediaPipe Hands setup
mp_hands = mp.solutions.hands
cap = cv2.VideoCapture(0)

canvas = None
previous_point = None
smoothed_point = None

strokes = []
current_stroke = []

MAX_UNDO = 10
undo_history = []

# Available colors (BGR format)
COLORS = [
    ("Green",  (0, 200, 0)),
    ("Red",    (0, 0, 255)),
    ("Blue",   (255, 0, 0)),
    ("Black",  (0, 0, 0)),
    ("Yellow", (0, 255, 255)),
    ("Purple", (180, 0, 180)),
    ("Orange", (0, 140, 255)),
    ("Pink",   (203, 192, 255))
]

selected_color_index = 0
brush_color = COLORS[selected_color_index][1]

# Finger detection
index_was_up = False
finger_up_count = 0
finger_down_count = 0

UP_THRESHOLD = 3
DOWN_THRESHOLD = 5

# Drawing settings
drawing_started = False
drawing_paused = False

# Lower value = smoother movement, but more delay
SMOOTHING = 0.25

BRUSH_THICKNESS = 6
CURSOR_RADIUS = 6
MIN_POINT_DISTANCE = 1.5

# Curve quality: more steps = smoother curves
CURVE_STEPS = 6


# Smooth hand movement using exponential smoothing
def smooth_point(new_point, old_point):
    if old_point is None:
        return new_point

    x = (
        SMOOTHING * new_point[0]
        + (1 - SMOOTHING) * old_point[0]
    )

    y = (
        SMOOTHING * new_point[1]
        + (1 - SMOOTHING) * old_point[1]
    )

    return (int(x), int(y))


# Catmull-Rom curve interpolation
def catmull_rom(p0, p1, p2, p3, steps=CURVE_STEPS):
    curve_points = []

    p0 = np.array(p0, dtype=np.float64)
    p1 = np.array(p1, dtype=np.float64)
    p2 = np.array(p2, dtype=np.float64)
    p3 = np.array(p3, dtype=np.float64)

    for i in range(steps):
        t = i / steps
        t2 = t * t
        t3 = t2 * t

        point = 0.5 * (
            2 * p1
            + (-p0 + p2) * t
            + (2*p0 - 5*p1 + 4*p2 - p3) * t2
            + (-p0 + 3*p1 - 3*p2 + p3) * t3
        )

        curve_points.append(
            (int(round(point[0])), int(round(point[1])))
        )

    return curve_points


# Convert stroke points into smooth curve points
def get_smooth_curve(points):
    if len(points) <= 2:
        return points.copy()

    result = []

    # Duplicate endpoints to keep the curve connected
    extended = [points[0]] + points + [points[-1]]

    for i in range(1, len(extended) - 2):
        segment = catmull_rom(
            extended[i - 1],
            extended[i],
            extended[i + 1],
            extended[i + 2]
        )

        result.extend(segment)

    result.append(points[-1])

    # Remove consecutive duplicate points
    clean_result = []

    for point in result:
        if not clean_result or point != clean_result[-1]:
            clean_result.append(point)

    return clean_result


# Save the current stroke
def save_current_stroke():
    global current_stroke, strokes, undo_history

    if current_stroke:
        stroke = {
            "points": current_stroke.copy(),
            "color": brush_color
        }

        strokes.append(stroke)
        undo_history.append(stroke)

        undo_history = undo_history[-MAX_UNDO:]
        current_stroke = []


# Draw a clean, smooth stroke
def draw_smooth_stroke(image, points, color):
    if not points:
        return

    radius = max(1, BRUSH_THICKNESS // 2)

    # Smooth curve through the collected points
    smooth_points = get_smooth_curve(points)

    if len(smooth_points) == 1:
        cv2.circle(
            image,
            smooth_points[0],
            radius,
            color,
            -1,
            cv2.LINE_AA
        )
        return

    # Draw anti-aliased curve segments
    for i in range(1, len(smooth_points)):
        cv2.line(
            image,
            smooth_points[i - 1],
            smooth_points[i],
            color,
            BRUSH_THICKNESS,
            cv2.LINE_AA
        )

    # Rounded joints prevent rough-looking corners
    for point in smooth_points:
        cv2.circle(
            image,
            point,
            radius,
            color,
            -1,
            cv2.LINE_AA
        )


# Small color palette
def draw_color_palette(image, selected_index):
    start_x = 20
    start_y = 23
    spacing = 32
    radius = 6

    cv2.rectangle(
        image,
        (8, 7),
        (start_x + 7 * spacing + 12, 48),
        (245, 245, 245),
        -1
    )

    for i, (name, color) in enumerate(COLORS):
        x = start_x + i * spacing
        y = start_y

        cv2.circle(
            image,
            (x, y),
            radius,
            color,
            -1,
            cv2.LINE_AA
        )

        if i == selected_index:
            cv2.circle(
                image,
                (x, y),
                radius + 2,
                (0, 0, 0),
                1,
                cv2.LINE_AA
            )

        cv2.putText(
            image,
            str(i + 1),
            (x - 3, y + 17),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.3,
            (0, 0, 0),
            1,
            cv2.LINE_AA
        )


# Start application
if not cap.isOpened():
    print("Camera open nahi ho raha.")

else:
    with mp_hands.Hands(
        static_image_mode=False,
        max_num_hands=1,
        model_complexity=0,
        min_detection_confidence=0.6,
        min_tracking_confidence=0.65
    ) as hands:

        while True:
            success, frame = cap.read()

            if not success:
                print("Camera frame read nahi ho raha.")
                break

            frame = cv2.flip(frame, 1)
            height, width, _ = frame.shape

            if canvas is None:
                canvas = np.full_like(frame, 255)

            # Hand detection
            rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            rgb.flags.writeable = False
            results = hands.process(rgb)

            status = "PRESS S TO START DRAWING"
            status_color = (180, 80, 0)
            cursor_point = None

            if drawing_started:

                if results.multi_hand_landmarks:
                    hand = results.multi_hand_landmarks[0]

                    tip = hand.landmark[8]
                    joint = hand.landmark[6]

                    x = int(tip.x * width)
                    y = int(tip.y * height)

                    smoothed_point = smooth_point(
                        (x, y),
                        smoothed_point
                    )

                    # Detect raised index finger
                    index_up_now = tip.y < joint.y

                    if index_up_now:
                        finger_up_count += 1
                        finger_down_count = 0
                    else:
                        finger_down_count += 1
                        finger_up_count = 0

                    if finger_up_count >= UP_THRESHOLD:
                        index_was_up = True

                    if finger_down_count >= DOWN_THRESHOLD:
                        index_was_up = False

                    if drawing_paused:
                        status = "PAUSED - P: Resume"
                        status_color = (0, 0, 255)

                        previous_point = None
                        save_current_stroke()

                    elif index_was_up:
                        status = "DRAWING"
                        status_color = brush_color
                        cursor_point = smoothed_point

                        if previous_point is None:
                            current_stroke = [smoothed_point]

                        else:
                            distance = math.hypot(
                                smoothed_point[0] - previous_point[0],
                                smoothed_point[1] - previous_point[1]
                            )

                            # Avoid unnecessary duplicate points
                            if distance >= MIN_POINT_DISTANCE:
                                current_stroke.append(smoothed_point)

                        previous_point = smoothed_point

                    else:
                        status = "HAND DETECTED - RAISE INDEX"
                        status_color = (0, 140, 255)

                        save_current_stroke()

                        previous_point = None
                        smoothed_point = None

                else:
                    status = (
                        "PAUSED - P: Resume"
                        if drawing_paused
                        else "NO HAND DETECTED"
                    )

                    status_color = (0, 0, 255)

                    save_current_stroke()

                    previous_point = None
                    smoothed_point = None

                    index_was_up = False
                    finger_up_count = 0
                    finger_down_count = 0

                if drawing_paused:
                    previous_point = None

            # White drawing canvas
            canvas = np.full_like(frame, 255)

            # Redraw saved strokes
            for stroke in strokes:
                draw_smooth_stroke(
                    canvas,
                    stroke["points"],
                    stroke["color"]
                )

            # Draw active stroke
            draw_smooth_stroke(
                canvas,
                current_stroke,
                brush_color
            )

            output = canvas.copy()

            # Display color palette
            draw_color_palette(output, selected_color_index)

            # Drawing cursor
            if cursor_point is not None and index_was_up:
                if drawing_started and not drawing_paused:
                    cv2.circle(
                        output,
                        cursor_point,
                        CURSOR_RADIUS,
                        brush_color,
                        -1,
                        cv2.LINE_AA
                    )

            # Status text
            cv2.putText(
                output,
                status,
                (20, 100),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                status_color,
                2,
                cv2.LINE_AA
            )

            # Instructions
            if not drawing_started:
                instructions = (
                    "S: Start | 1-8: Color | C: Clear | Q: Quit"
                )
            else:
                instructions = (
                    "1-8: Color | P: Pause/Resume | "
                    "Z: Undo | C: Clear | Q: Quit"
                )

            cv2.putText(
                output,
                instructions,
                (20, height - 20),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (80, 80, 80),
                1,
                cv2.LINE_AA
            )

            cv2.imshow("Virtual Paint App", output)

            key = cv2.waitKey(1) & 0xFF

            # Select color
            if ord("1") <= key <= ord("8"):
                selected_color_index = key - ord("1")
                brush_color = COLORS[selected_color_index][1]

            # Start drawing
            elif key == ord("s") and not drawing_started:
                drawing_started = True
                drawing_paused = False

                previous_point = None
                smoothed_point = None
                current_stroke = []

                index_was_up = False
                finger_up_count = 0
                finger_down_count = 0

            # Pause / resume
            elif key == ord("p") and drawing_started:
                drawing_paused = not drawing_paused

                previous_point = None
                smoothed_point = None

                save_current_stroke()

                index_was_up = False
                finger_up_count = 0
                finger_down_count = 0

            # Undo
            elif key == ord("z") and drawing_started:
                if current_stroke:
                    save_current_stroke()

                if undo_history:
                    last_stroke = undo_history.pop()

                    for i in range(len(strokes) - 1, -1, -1):
                        if strokes[i] is last_stroke:
                            strokes.pop(i)
                            break

                previous_point = None
                smoothed_point = None

                index_was_up = False
                finger_up_count = 0
                finger_down_count = 0

            # Clear canvas
            elif key == ord("c"):
                strokes = []
                current_stroke = []
                undo_history = []

                canvas = np.full_like(frame, 255)

                drawing_started = False
                drawing_paused = False

                previous_point = None
                smoothed_point = None

                index_was_up = False
                finger_up_count = 0
                finger_down_count = 0

            # Quit
            elif key == ord("q"):
                break

    cap.release()
    cv2.destroyAllWindows()
```

```python

import cv2
import mediapipe as mp
import numpy as np
import math

# ==========================================
# 1. CAMERA SETUP
# ==========================================

cap = cv2.VideoCapture(0)

cap.set(cv2.CAP_PROP_FRAME_WIDTH, 1280)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 720)

mp_hands = mp.solutions.hands

canvas = None
previous_point = None
smoothed_point = None

strokes = []
current_stroke = []

undo_history = []
MAX_UNDO = 10


# ==========================================
# 2. COLORS
# ==========================================

COLORS = [
    ("Green",  (0, 200, 0)),
    ("Red",    (0, 0, 255)),
    ("Blue",   (255, 0, 0)),
    ("Black",  (0, 0, 0)),
    ("Yellow", (0, 255, 255)),
    ("Purple", (180, 0, 180)),
    ("Orange", (0, 140, 255)),
    ("Pink",   (203, 192, 255))
]

selected_color_index = 0
brush_color = COLORS[selected_color_index][1]
current_stroke_color = brush_color


# ==========================================
# 3. FINGER DETECTION SETTINGS
# ==========================================

index_was_up = False

finger_up_count = 0
finger_down_count = 0

UP_THRESHOLD = 3
DOWN_THRESHOLD = 5


# ==========================================
# 4. DRAWING SETTINGS
# ==========================================

drawing_started = False
drawing_paused = False

SMOOTHING = 0.20

BRUSH_THICKNESS = 6
CURSOR_RADIUS = 6

MIN_POINT_DISTANCE = 1.0
CURVE_STEPS = 6


# Side-angle finger detection settings
FINGER_EXTENSION_ANGLE = 145
FINGER_EXTENSION_RATIO = 1.10


# ==========================================
# 5. 3D DISTANCE FUNCTION
# ==========================================

def distance_3d(point1, point2):

    return math.sqrt(
        (point1.x - point2.x) ** 2
        + (point1.y - point2.y) ** 2
        + (point1.z - point2.z) ** 2
    )


# ==========================================
# 6. DETECT INDEX FINGER EXTENSION
# ==========================================

def is_index_finger_extended(hand):

    # Index finger landmarks
    mcp = hand.landmark[5]
    pip = hand.landmark[6]
    tip = hand.landmark[8]

    # Vectors from PIP joint
    vector1 = np.array([
        mcp.x - pip.x,
        mcp.y - pip.y,
        mcp.z - pip.z
    ])

    vector2 = np.array([
        tip.x - pip.x,
        tip.y - pip.y,
        tip.z - pip.z
    ])

    # Calculate finger bend angle
    denominator = (
        np.linalg.norm(vector1)
        * np.linalg.norm(vector2)
    )

    if denominator < 1e-8:
        return False

    cosine_angle = np.dot(vector1, vector2) / denominator

    cosine_angle = np.clip(
        cosine_angle, -1.0, 1.0
    )

    angle = math.degrees(
        math.acos(cosine_angle)
    )

    # Check finger extension
    mcp_to_tip = distance_3d(mcp, tip)
    mcp_to_pip = distance_3d(mcp, pip)

    extended = (
        angle >= FINGER_EXTENSION_ANGLE
        and mcp_to_tip >= mcp_to_pip * FINGER_EXTENSION_RATIO
    )

    return extended


# ==========================================
# 7. HAND MOVEMENT SMOOTHING
# ==========================================

def smooth_point(new_point, old_point):

    if old_point is None:
        return new_point

    x = (
        SMOOTHING * new_point[0]
        + (1 - SMOOTHING) * old_point[0]
    )

    y = (
        SMOOTHING * new_point[1]
        + (1 - SMOOTHING) * old_point[1]
    )

    return (
        int(round(x)),
        int(round(y))
    )


# ==========================================
# 8. CATMULL-ROM CURVE
# ==========================================

def catmull_rom(p0, p1, p2, p3, steps=CURVE_STEPS):

    curve_points = []

    p0 = np.array(p0, dtype=np.float64)
    p1 = np.array(p1, dtype=np.float64)
    p2 = np.array(p2, dtype=np.float64)
    p3 = np.array(p3, dtype=np.float64)

    for i in range(steps):

        t = i / steps
        t2 = t * t
        t3 = t2 * t

        point = 0.5 * (
            2 * p1
            + (-p0 + p2) * t
            + (2*p0 - 5*p1 + 4*p2 - p3) * t2
            + (-p0 + 3*p1 - 3*p2 + p3) * t3
        )

        curve_points.append((
            int(round(point[0])),
            int(round(point[1]))
        ))

    return curve_points


def get_smooth_curve(points):

    if len(points) <= 2:
        return points.copy()

    result = []

    extended = [points[0]] + points + [points[-1]]

    for i in range(1, len(extended) - 2):

        segment = catmull_rom(
            extended[i - 1],
            extended[i],
            extended[i + 1],
            extended[i + 2]
        )

        result.extend(segment)

    result.append(points[-1])

    clean_result = []

    for point in result:

        if not clean_result or point != clean_result[-1]:
            clean_result.append(point)

    return clean_result


# ==========================================
# 9. DRAW SMOOTH STROKE
# ==========================================

def draw_smooth_stroke(image, points, color):

    if not points:
        return

    radius = max(1, BRUSH_THICKNESS // 2)

    smooth_points = get_smooth_curve(points)

    if len(smooth_points) == 1:

        cv2.circle(
            image,
            smooth_points[0],
            radius,
            color,
            -1,
            cv2.LINE_AA
        )

        return

    for i in range(1, len(smooth_points)):

        cv2.line(
            image,
            smooth_points[i - 1],
            smooth_points[i],
            color,
            BRUSH_THICKNESS,
            cv2.LINE_AA
        )

    for point in smooth_points:

        cv2.circle(
            image,
            point,
            radius,
            color,
            -1,
            cv2.LINE_AA
        )


# ==========================================
# 10. SAVE CURRENT STROKE
# ==========================================

def save_current_stroke():

    global current_stroke
    global strokes
    global undo_history
    global canvas

    if current_stroke:

        stroke = {
            "points": current_stroke.copy(),
            "color": current_stroke_color
        }

        strokes.append(stroke)
        undo_history.append(stroke)

        undo_history = undo_history[-MAX_UNDO:]

        draw_smooth_stroke(
            canvas,
            stroke["points"],
            stroke["color"]
        )

        current_stroke = []


# ==========================================
# 11. REBUILD CANVAS AFTER UNDO
# ==========================================

def rebuild_canvas():

    global canvas

    canvas[:] = 255

    for stroke in strokes:

        draw_smooth_stroke(
            canvas,
            stroke["points"],
            stroke["color"]
        )


# ==========================================
# 12. SMALL COLOR PALETTE
# ==========================================

def draw_color_palette(image, selected_index):

    start_x = 20
    start_y = 23
    spacing = 32
    radius = 6

    cv2.rectangle(
        image,
        (8, 7),
        (start_x + 7 * spacing + 12, 48),
        (245, 245, 245),
        -1
    )

    for i, (name, color) in enumerate(COLORS):

        x = start_x + i * spacing
        y = start_y

        cv2.circle(
            image,
            (x, y),
            radius,
            color,
            -1,
            cv2.LINE_AA
        )

        if i == selected_index:

            cv2.circle(
                image,
                (x, y),
                radius + 2,
                (0, 0, 0),
                1,
                cv2.LINE_AA
            )

        cv2.putText(
            image,
            str(i + 1),
            (x - 3, y + 17),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.3,
            (0, 0, 0),
            1,
            cv2.LINE_AA
        )


# ==========================================
# 13. START APPLICATION
# ==========================================

if not cap.isOpened():

    print("Camera open nahi ho raha.")

else:

    with mp_hands.Hands(

        static_image_mode=False,
        max_num_hands=1,

        model_complexity=1,

        min_detection_confidence=0.4,
        min_tracking_confidence=0.5

    ) as hands:

        while True:

            success, frame = cap.read()

            if not success:

                print("Camera frame read nahi ho raha.")
                break

            frame = cv2.flip(frame, 1)

            height, width, _ = frame.shape

            if canvas is None:

                canvas = np.full_like(frame, 255)

            # ==================================
            # 14. HAND DETECTION
            # ==================================

            rgb = cv2.cvtColor(
                frame,
                cv2.COLOR_BGR2RGB
            )

            rgb.flags.writeable = False

            results = hands.process(rgb)

            status = "PRESS S TO START DRAWING"
            status_color = (180, 80, 0)

            cursor_point = None

            if drawing_started:

                if results.multi_hand_landmarks:

                    hand = results.multi_hand_landmarks[0]

                    tip = hand.landmark[8]

                    x = int(tip.x * width)
                    y = int(tip.y * height)

                    smoothed_point = smooth_point(
                        (x, y),
                        smoothed_point
                    )

                    # New angle-independent finger detection
                    index_up_now = is_index_finger_extended(hand)

                    # Stabilize detection
                    if index_up_now:

                        finger_up_count += 1
                        finger_down_count = 0

                    else:

                        finger_down_count += 1
                        finger_up_count = 0

                    if finger_up_count >= UP_THRESHOLD:

                        index_was_up = True

                    if finger_down_count >= DOWN_THRESHOLD:

                        index_was_up = False

                    # ==================================
                    # 15. DRAWING CONTROL
                    # ==================================

                    if drawing_paused:

                        status = "PAUSED - P: Resume"
                        status_color = (0, 0, 255)

                        previous_point = None

                        save_current_stroke()

                    elif index_was_up:

                        status = "DRAWING"
                        status_color = brush_color

                        cursor_point = smoothed_point

                        if previous_point is None:

                            current_stroke = [smoothed_point]

                            current_stroke_color = brush_color

                        else:

                            distance = math.hypot(
                                smoothed_point[0] - previous_point[0],
                                smoothed_point[1] - previous_point[1]
                            )

                            if distance >= MIN_POINT_DISTANCE:

                                current_stroke.append(smoothed_point)

                        previous_point = smoothed_point

                    else:

                        status = "HAND DETECTED - EXTEND INDEX"
                        status_color = (0, 140, 255)

                        save_current_stroke()

                        previous_point = None
                        smoothed_point = None

                else:

                    status = (
                        "PAUSED - P: Resume"
                        if drawing_paused
                        else "NO HAND DETECTED"
                    )

                    status_color = (0, 0, 255)

                    save_current_stroke()

                    previous_point = None
                    smoothed_point = None

                    index_was_up = False

                    finger_up_count = 0
                    finger_down_count = 0

                if drawing_paused:

                    previous_point = None

            # ==================================
            # 16. FAST CANVAS RENDERING
            # ==================================

            output = canvas.copy()

            # Draw active stroke
            if current_stroke:

                draw_smooth_stroke(
                    output,
                    current_stroke,
                    current_stroke_color
                )

            draw_color_palette(
                output,
                selected_color_index
            )

            # Cursor
            if cursor_point is not None and index_was_up:

                if drawing_started and not drawing_paused:

                    cv2.circle(
                        output,
                        cursor_point,
                        CURSOR_RADIUS,
                        brush_color,
                        -1,
                        cv2.LINE_AA
                    )

            # Status text
            cv2.putText(
                output,
                status,
                (20, 100),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                status_color,
                2,
                cv2.LINE_AA
            )

            # Instructions
            if not drawing_started:

                instructions = (
                    "S: Start | 1-8: Color | C: Clear | Q: Quit"
                )

            else:

                instructions = (
                    "1-8: Color | P: Pause/Resume | "
                    "Z: Undo | C: Clear | Q: Quit"
                )

            cv2.putText(
                output,
                instructions,
                (20, height - 20),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (80, 80, 80),
                1,
                cv2.LINE_AA
            )

            cv2.imshow(
                "Virtual Paint App",
                output
            )

            key = cv2.waitKey(1) & 0xFF

            # ==================================
            # 17. KEYBOARD CONTROLS
            # ==================================

            # Select color
            if ord("1") <= key <= ord("8"):

                selected_color_index = key - ord("1")

                brush_color = COLORS[selected_color_index][1]

            # Start drawing
            elif key == ord("s") and not drawing_started:

                drawing_started = True
                drawing_paused = False

                previous_point = None
                smoothed_point = None

                current_stroke = []

                index_was_up = False

                finger_up_count = 0
                finger_down_count = 0

            # Pause / Resume
            elif key == ord("p") and drawing_started:

                drawing_paused = not drawing_paused

                previous_point = None
                smoothed_point = None

                save_current_stroke()

                index_was_up = False

                finger_up_count = 0
                finger_down_count = 0

            # Undo
            elif key == ord("z") and drawing_started:

                if current_stroke:

                    save_current_stroke()

                if undo_history:

                    last_stroke = undo_history.pop()

                    for i in range(len(strokes) - 1, -1, -1):

                        if strokes[i] is last_stroke:

                            strokes.pop(i)
                            break

                    rebuild_canvas()

                previous_point = None
                smoothed_point = None

                index_was_up = False

                finger_up_count = 0
                finger_down_count = 0

            # Clear
            elif key == ord("c"):

                strokes = []
                current_stroke = []
                undo_history = []

                canvas = np.full_like(frame, 255)

                drawing_started = False
                drawing_paused = False

                previous_point = None
                smoothed_point = None

                index_was_up = False

                finger_up_count = 0
                finger_down_count = 0

            # Quit
            elif key == ord("q"):

                break

    cap.release()
    cv2.destroyAllWindows()
```

```python

import cv2
import mediapipe as mp
import numpy as np
import math

# ============================================================
# 1. CAMERA SETUP
# ============================================================

CAMERA_INDEX = 0
FRAME_WIDTH = 1280
FRAME_HEIGHT = 720

cap = cv2.VideoCapture(CAMERA_INDEX)

if not cap.isOpened():
    cap.release()
    cap = cv2.VideoCapture(1)

if not cap.isOpened():
    raise RuntimeError("Camera not found. Check your camera connection.")

cap.set(cv2.CAP_PROP_FRAME_WIDTH, FRAME_WIDTH)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, FRAME_HEIGHT)
cap.set(cv2.CAP_PROP_FPS, 30)
cap.set(cv2.CAP_PROP_BUFFERSIZE, 1)

mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils

# ============================================================
# 2. COLORS
# ============================================================

COLORS = [
    ("Green",  (0, 200, 0)),
    ("Red",    (0, 0, 255)),
    ("Blue",   (255, 0, 0)),
    ("Black",  (0, 0, 0)),
    ("Yellow", (0, 255, 255)),
    ("Purple", (180, 0, 180)),
    ("Orange", (0, 140, 255)),
    ("Pink",   (203, 192, 255))
]

selected_color_index = 0
brush_color = COLORS[selected_color_index][1]

# ============================================================
# 3. DRAWING TOOLS
# ============================================================

TOOLS = {
    "PEN": {"thickness": 2, "opacity": 1.0},
    "MARKER": {"thickness": 9, "opacity": 0.55},
    "MAX THICKNESS": {"thickness": 24, "opacity": 1.0},
    "PENCIL": {"thickness": 2, "opacity": 0.80},
    "BRUSH": {"thickness": 15, "opacity": 0.90}
}

selected_tool = "PEN"
current_stroke_tool = selected_tool
current_stroke_color = brush_color

# ============================================================
# 4. CANVAS AND APP VARIABLES
# ============================================================

canvas = None

strokes = []
current_stroke = []
undo_history = []

MAX_UNDO = 10

drawing_started = False
drawing_paused = False

previous_point = None
smoothed_point = None

# ============================================================
# 5. HAND DETECTION SETTINGS
# ============================================================

PROCESS_WIDTH = 640
PROCESS_HEIGHT = 360

MIN_DETECTION_CONFIDENCE = 0.50
MIN_TRACKING_CONFIDENCE = 0.50

UP_THRESHOLD = 2
DOWN_THRESHOLD = 3

finger_up_count = 0
finger_down_count = 0
index_is_extended = False

lost_hand_frames = 0
MAX_LOST_HAND_FRAMES = 2

# ============================================================
# 6. DRAWING SETTINGS
# ============================================================

SMOOTHING = 0.38
MIN_POINT_DISTANCE = 1.2
CURVE_STEPS = 6
CURSOR_RADIUS = 6

# More tolerant finger detection
ANGLE_THRESHOLD_2D = 125
ANGLE_THRESHOLD_3D = 130
LENGTH_RATIO_THRESHOLD = 1.00

SHOW_DEBUG_WINDOW = True

# ============================================================
# 7. DISTANCE FUNCTIONS
# ============================================================

def distance_2d(a, b):
    return math.hypot(a.x - b.x, a.y - b.y)


def distance_3d(a, b):
    return math.sqrt(
        (a.x - b.x) ** 2 +
        (a.y - b.y) ** 2 +
        (a.z - b.z) ** 2
    )


# ============================================================
# 8. ANGLE FUNCTIONS
# ============================================================

def calculate_angle_2d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


def calculate_angle_3d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y, a.z - b.z],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y, c.z - b.z],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


# ============================================================
# 9. IMPROVED INDEX FINGER DETECTION
# ============================================================

def is_index_finger_extended(hand):

    lm = hand.landmark

    mcp = lm[5]
    pip = lm[6]
    dip = lm[7]
    tip = lm[8]

    angle_2d = calculate_angle_2d(mcp, pip, tip)
    angle_3d = calculate_angle_3d(mcp, pip, tip)

    dip_angle_2d = calculate_angle_2d(pip, dip, tip)
    dip_angle_3d = calculate_angle_3d(pip, dip, tip)

    ratio_2d = (
        distance_2d(mcp, tip) /
        max(distance_2d(mcp, pip), 0.0001)
    )

    ratio_3d = (
        distance_3d(mcp, tip) /
        max(distance_3d(mcp, pip), 0.0001)
    )

    angle_ok = (
        angle_2d >= ANGLE_THRESHOLD_2D or
        angle_3d >= ANGLE_THRESHOLD_3D
    )

    length_ok = (
        ratio_2d >= LENGTH_RATIO_THRESHOLD or
        ratio_3d >= LENGTH_RATIO_THRESHOLD
    )

    tip_ok = (
        dip_angle_2d >= 110 or
        dip_angle_3d >= 110
    )

    return angle_ok and length_ok and tip_ok


# ============================================================
# 10. GESTURE STABILIZATION
# ============================================================

def update_gesture(raw_extended):

    global finger_up_count
    global finger_down_count
    global index_is_extended

    if raw_extended:

        finger_up_count += 1
        finger_down_count = 0

    else:

        finger_down_count += 1
        finger_up_count = 0

    if finger_up_count >= UP_THRESHOLD:
        index_is_extended = True

    elif finger_down_count >= DOWN_THRESHOLD:
        index_is_extended = False

    return index_is_extended


def reset_tracking():

    global previous_point
    global smoothed_point
    global finger_up_count
    global finger_down_count
    global index_is_extended

    previous_point = None
    smoothed_point = None

    finger_up_count = 0
    finger_down_count = 0

    index_is_extended = False


# ============================================================
# 11. CURSOR SMOOTHING
# ============================================================

def smooth_point(new_point, old_point):

    if old_point is None:
        return new_point

    x = (
        SMOOTHING * new_point[0] +
        (1 - SMOOTHING) * old_point[0]
    )

    y = (
        SMOOTHING * new_point[1] +
        (1 - SMOOTHING) * old_point[1]
    )

    return int(x), int(y)


# ============================================================
# 12. CATMULL-ROM CURVE
# ============================================================

def catmull_rom(p0, p1, p2, p3):

    result = []

    p0 = np.array(p0, dtype=np.float64)
    p1 = np.array(p1, dtype=np.float64)
    p2 = np.array(p2, dtype=np.float64)
    p3 = np.array(p3, dtype=np.float64)

    for i in range(CURVE_STEPS):

        t = i / CURVE_STEPS
        t2 = t * t
        t3 = t2 * t

        point = 0.5 * (
            2 * p1
            + (-p0 + p2) * t
            + (2*p0 - 5*p1 + 4*p2 - p3) * t2
            + (-p0 + 3*p1 - 3*p2 + p3) * t3
        )

        result.append((
            int(round(point[0])),
            int(round(point[1]))
        ))

    return result


def get_smooth_curve(points):

    if len(points) <= 2:
        return points.copy()

    result = []

    extended = [points[0]] + points + [points[-1]]

    for i in range(1, len(extended) - 2):

        result.extend(
            catmull_rom(
                extended[i - 1],
                extended[i],
                extended[i + 1],
                extended[i + 2]
            )
        )

    result.append(points[-1])

    return result


# ============================================================
# 13. DRAWING FUNCTION
# ============================================================

def draw_smooth_stroke(image, points, color, tool="PEN"):

    if not points:
        return

    settings = TOOLS.get(tool, TOOLS["PEN"])

    thickness = settings["thickness"]
    opacity = settings["opacity"]

    radius = max(1, thickness // 2)

    smooth_points = get_smooth_curve(points)

    # Separate layer prevents brush blur from damaging old strokes
    layer = np.zeros_like(image)

    if len(smooth_points) == 1:

        cv2.circle(
            layer,
            smooth_points[0],
            radius,
            color,
            -1,
            cv2.LINE_AA
        )

    else:

        pts = np.array(
            smooth_points,
            dtype=np.int32
        ).reshape((-1, 1, 2))

        draw_color = color

        if tool == "PENCIL":
            draw_color = tuple(
                int(c * 0.85 + 255 * 0.15)
                for c in color
            )

        cv2.polylines(
            layer,
            [pts],
            False,
            draw_color,
            thickness,
            cv2.LINE_AA
        )

        cv2.circle(
            layer,
            smooth_points[0],
            radius,
            draw_color,
            -1,
            cv2.LINE_AA
        )

        cv2.circle(
            layer,
            smooth_points[-1],
            radius,
            draw_color,
            -1,
            cv2.LINE_AA
        )

    if tool == "PENCIL" and len(smooth_points) == 1:
        pencil_color = tuple(
            int(c * 0.85 + 255 * 0.15)
            for c in color
        )
        layer[:] = 0
        cv2.circle(
            layer,
            smooth_points[0],
            radius,
            pencil_color,
            -1,
            cv2.LINE_AA
        )

    elif tool == "BRUSH":

        layer = cv2.GaussianBlur(
            layer,
            (5, 5),
            1.0
        )

    mask = cv2.cvtColor(
        layer,
        cv2.COLOR_BGR2GRAY
    )

    active = mask > 0

    if not np.any(active):
        return

    if opacity >= 1.0:

        image[active] = layer[active]

    else:

        blended = (
            layer.astype(np.float32) * opacity +
            image.astype(np.float32) * (1 - opacity)
        )

        image[active] = np.clip(
            blended[active],
            0,
            255
        ).astype(np.uint8)


# ============================================================
# 14. SAVE STROKE
# ============================================================

def save_current_stroke():

    global current_stroke
    global strokes
    global undo_history

    if not current_stroke:
        return

    stroke = {
        "points": current_stroke.copy(),
        "color": current_stroke_color,
        "tool": current_stroke_tool
    }

    strokes.append(stroke)

    undo_history.append(stroke)
    undo_history = undo_history[-MAX_UNDO:]

    draw_smooth_stroke(
        canvas,
        stroke["points"],
        stroke["color"],
        stroke["tool"]
    )

    current_stroke = []


# ============================================================
# 15. REBUILD CANVAS
# ============================================================

def rebuild_canvas():

    canvas[:] = 255

    for stroke in strokes:

        draw_smooth_stroke(
            canvas,
            stroke["points"],
            stroke["color"],
            stroke["tool"]
        )


# ============================================================
# 16. COLOR PALETTE
# ============================================================

def draw_color_palette(image, selected_index):

    start_x = 20
    start_y = 23

    spacing = 32
    radius = 6

    cv2.rectangle(
        image,
        (8, 7),
        (start_x + 7 * spacing + 12, 48),
        (245, 245, 245),
        -1
    )

    for i, (name, color) in enumerate(COLORS):

        x = start_x + i * spacing
        y = start_y

        cv2.circle(
            image,
            (x, y),
            radius,
            color,
            -1,
            cv2.LINE_AA
        )

        if i == selected_index:

            cv2.circle(
                image,
                (x, y),
                radius + 3,
                (0, 0, 0),
                1,
                cv2.LINE_AA
            )

        cv2.putText(
            image,
            str(i + 1),
            (x - 3, y + 17),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.3,
            (0, 0, 0),
            1,
            cv2.LINE_AA
        )


# ============================================================
# 17. TOOL STATUS
# ============================================================

def draw_tool_status(image):

    text = (
        f"TOOL: {selected_tool} | "
        f"SIZE: {TOOLS[selected_tool]['thickness']}"
    )

    cv2.putText(
        image,
        text,
        (20, 135),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.55,
        (40, 40, 40),
        2,
        cv2.LINE_AA
    )


# ============================================================
# 18. MAIN APPLICATION
# ============================================================

try:

    with mp_hands.Hands(

        static_image_mode=False,

        max_num_hands=1,

        model_complexity=1,

        min_detection_confidence=MIN_DETECTION_CONFIDENCE,

        min_tracking_confidence=MIN_TRACKING_CONFIDENCE

    ) as hands:

        while True:

            success, frame = cap.read()

            if not success:
                print("Unable to read camera frame.")
                break

            frame = cv2.flip(frame, 1)

            height, width = frame.shape[:2]

            if canvas is None:
                canvas = np.full_like(frame, 255)

            # ----------------------------------------------
            # CONTINUOUS HAND DETECTION
            # ----------------------------------------------

            small_frame = cv2.resize(
                frame,
                (PROCESS_WIDTH, PROCESS_HEIGHT),
                interpolation=cv2.INTER_AREA
            )

            rgb = cv2.cvtColor(
                small_frame,
                cv2.COLOR_BGR2RGB
            )

            rgb.flags.writeable = False
            results = hands.process(rgb)
            rgb.flags.writeable = True

            hand_detected = bool(results.multi_hand_landmarks)

            status = "NO HAND DETECTED - SHOW YOUR HAND"
            status_color = (0, 0, 255)

            cursor_point = None

            # Debug window always displays camera tracking result
            debug_frame = small_frame.copy()

            if hand_detected:

                hand = results.multi_hand_landmarks[0]

                mp_drawing.draw_landmarks(
                    debug_frame,
                    hand,
                    mp_hands.HAND_CONNECTIONS
                )

                tip = hand.landmark[8]

                x = int(tip.x * width)
                y = int(tip.y * height)

                x = max(0, min(width - 1, x))
                y = max(0, min(height - 1, y))

                smoothed_point = smooth_point(
                    (x, y),
                    smoothed_point
                )

                cursor_point = smoothed_point

                lost_hand_frames = 0

                raw_extended = is_index_finger_extended(hand)

                index_is_extended = update_gesture(raw_extended)

                if not drawing_started:

                    status = "HAND DETECTED - PRESS S TO START"
                    status_color = (0, 150, 0)

                elif drawing_paused:

                    status = "PAUSED - PRESS P TO RESUME"
                    status_color = (0, 0, 255)

                    save_current_stroke()
                    previous_point = None

                elif index_is_extended:

                    status = "DRAWING"
                    status_color = brush_color

                    if previous_point is None:

                        current_stroke = [smoothed_point]

                        current_stroke_color = brush_color
                        current_stroke_tool = selected_tool

                    else:

                        movement = math.hypot(
                            smoothed_point[0] - previous_point[0],
                            smoothed_point[1] - previous_point[1]
                        )

                        if movement >= MIN_POINT_DISTANCE:
                            current_stroke.append(smoothed_point)

                    previous_point = smoothed_point

                else:

                    status = "HAND DETECTED - EXTEND INDEX FINGER"
                    status_color = (0, 140, 255)

                    save_current_stroke()
                    previous_point = None

            else:

                lost_hand_frames += 1

                if lost_hand_frames >= MAX_LOST_HAND_FRAMES:

                    save_current_stroke()
                    reset_tracking()

            # ----------------------------------------------
            # DEBUG WINDOW
            # ----------------------------------------------

            debug_text = (
                "HAND FOUND" if hand_detected else "NO HAND"
            )

            debug_color = (
                (0, 255, 0) if hand_detected else (0, 0, 255)
            )

            cv2.putText(
                debug_frame,
                debug_text,
                (10, 25),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                debug_color,
                2,
                cv2.LINE_AA
            )

            if SHOW_DEBUG_WINDOW:
                cv2.imshow("Hand Tracking Debug", debug_frame)

            # ----------------------------------------------
            # CANVAS DISPLAY
            # ----------------------------------------------

            output = canvas.copy()

            if current_stroke:

                draw_smooth_stroke(
                    output,
                    current_stroke,
                    current_stroke_color,
                    current_stroke_tool
                )

            draw_color_palette(
                output,
                selected_color_index
            )

            draw_tool_status(output)

            if (
                cursor_point is not None
                and drawing_started
                and not drawing_paused
            ):

                cv2.circle(
                    output,
                    cursor_point,
                    max(3, TOOLS[selected_tool]["thickness"] // 2),
                    brush_color,
                    -1,
                    cv2.LINE_AA
                )

            cv2.putText(
                output,
                status,
                (20, 100),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                status_color,
                2,
                cv2.LINE_AA
            )

            instructions = (
                "S: Start | 1-8: Color | N: Pen | M: Marker | "
                "X: Max | L: Pencil | B: Brush | "
                "P: Pause | Z: Undo | C: Clear | Q: Quit"
            )

            cv2.putText(
                output,
                instructions,
                (20, height - 20),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.43,
                (80, 80, 80),
                1,
                cv2.LINE_AA
            )

            cv2.imshow("Virtual Paint App", output)

            # ----------------------------------------------
            # KEYBOARD CONTROLS
            # ----------------------------------------------

            key = cv2.waitKey(1) & 0xFF

            if ord("1") <= key <= ord("8"):

                selected_color_index = key - ord("1")
                brush_color = COLORS[selected_color_index][1]

            elif key == ord("n"):
                selected_tool = "PEN"

            elif key == ord("m"):
                selected_tool = "MARKER"

            elif key == ord("x"):
                selected_tool = "MAX THICKNESS"

            elif key == ord("l"):
                selected_tool = "PENCIL"

            elif key == ord("b"):
                selected_tool = "BRUSH"

            elif key == ord("s") and not drawing_started:

                drawing_started = True
                drawing_paused = False

                reset_tracking()
                save_current_stroke()

            elif key == ord("p") and drawing_started:

                drawing_paused = not drawing_paused

                save_current_stroke()
                reset_tracking()

            elif key == ord("z") and drawing_started:

                save_current_stroke()

                if undo_history:

                    last_stroke = undo_history.pop()

                    for i in range(len(strokes) - 1, -1, -1):

                        if strokes[i] is last_stroke:
                            strokes.pop(i)
                            break

                    rebuild_canvas()

                reset_tracking()

            elif key == ord("c"):

                strokes.clear()
                current_stroke.clear()
                undo_history.clear()

                canvas[:] = 255

                drawing_started = False
                drawing_paused = False

                lost_hand_frames = 0
                reset_tracking()

            elif key == ord("q"):

                break

finally:

    cap.release()
    cv2.destroyAllWindows()
```

```python
import cv2
import mediapipe as mp
import numpy as np
import math

# ============================================================
# 1. CAMERA SETUP
# ============================================================

CAMERA_INDEX = 0
FRAME_WIDTH = 1280
FRAME_HEIGHT = 720

cap = cv2.VideoCapture(CAMERA_INDEX)

if not cap.isOpened():
    cap.release()
    cap = cv2.VideoCapture(1)

if not cap.isOpened():
    raise RuntimeError("Camera not found. Check your camera connection.")

cap.set(cv2.CAP_PROP_FRAME_WIDTH, FRAME_WIDTH)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, FRAME_HEIGHT)
cap.set(cv2.CAP_PROP_FPS, 30)
cap.set(cv2.CAP_PROP_BUFFERSIZE, 1)

mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils

# ============================================================
# 2. COLORS
# ============================================================

COLORS = [
    ("Green",  (0, 200, 0)),
    ("Red",    (0, 0, 255)),
    ("Blue",   (255, 0, 0)),
    ("Black",  (0, 0, 0)),
    ("Yellow", (0, 255, 255)),
    ("Purple", (180, 0, 180)),
    ("Orange", (0, 140, 255)),
    ("Pink",   (203, 192, 255))
]

selected_color_index = 0
brush_color = COLORS[selected_color_index][1]

# ============================================================
# 3. DRAWING TOOLS
# ============================================================

TOOLS = {
    "PEN": {"thickness": 2, "opacity": 1.0},
    "MARKER": {"thickness": 9, "opacity": 0.55},
    "MAX THICKNESS": {"thickness": 24, "opacity": 1.0},
    "PENCIL": {"thickness": 2, "opacity": 0.80},
    "BRUSH": {"thickness": 15, "opacity": 0.90}
}

selected_tool = "PEN"
current_stroke_tool = selected_tool
current_stroke_color = brush_color

# ============================================================
# 4. CANVAS AND APP VARIABLES
# ============================================================

canvas = None

strokes = []
current_stroke = []
undo_history = []

MAX_UNDO = 10

drawing_started = False
drawing_paused = False

previous_point = None
smoothed_point = None

# ============================================================
# 5. HAND DETECTION SETTINGS
# ============================================================

PROCESS_WIDTH = 640
PROCESS_HEIGHT = 360

MIN_DETECTION_CONFIDENCE = 0.50
MIN_TRACKING_CONFIDENCE = 0.50

UP_THRESHOLD = 2
DOWN_THRESHOLD = 3

finger_up_count = 0
finger_down_count = 0
index_is_extended = False

lost_hand_frames = 0
MAX_LOST_HAND_FRAMES = 2

# ============================================================
# 6. IMPROVED DRAWING SETTINGS
# ============================================================

# Cursor smoothing
SMOOTHING = 0.28

# Minimum distance between stored drawing points
MIN_POINT_DISTANCE = 0.8

# More curve steps = smoother curves
CURVE_STEPS = 10

CURSOR_RADIUS = 6

# Stationary finger protection
MOVEMENT_START_THRESHOLD = 4.5
STATIONARY_RADIUS = 2.5
STATIONARY_FRAMES = 8

# Small raw coordinate history for stationary detection
raw_point_history = []

stationary_anchor = None
filtered_point = None
movement_active = False

SHOW_DEBUG_WINDOW = True

# More tolerant finger detection
ANGLE_THRESHOLD_2D = 125
ANGLE_THRESHOLD_3D = 130
LENGTH_RATIO_THRESHOLD = 1.00

# ============================================================
# 7. DISTANCE FUNCTIONS
# ============================================================

def distance_2d(a, b):
    return math.hypot(a.x - b.x, a.y - b.y)


def distance_3d(a, b):
    return math.sqrt(
        (a.x - b.x) ** 2 +
        (a.y - b.y) ** 2 +
        (a.z - b.z) ** 2
    )


def point_distance(p1, p2):
    return math.hypot(
        p1[0] - p2[0],
        p1[1] - p2[1]
    )


# ============================================================
# 8. ANGLE FUNCTIONS
# ============================================================

def calculate_angle_2d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


def calculate_angle_3d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y, a.z - b.z],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y, c.z - b.z],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


# ============================================================
# 9. INDEX FINGER DETECTION
# ============================================================

def is_index_finger_extended(hand):

    lm = hand.landmark

    mcp = lm[5]
    pip = lm[6]
    dip = lm[7]
    tip = lm[8]

    angle_2d = calculate_angle_2d(mcp, pip, tip)
    angle_3d = calculate_angle_3d(mcp, pip, tip)

    dip_angle_2d = calculate_angle_2d(pip, dip, tip)
    dip_angle_3d = calculate_angle_3d(pip, dip, tip)

    ratio_2d = (
        distance_2d(mcp, tip) /
        max(distance_2d(mcp, pip), 0.0001)
    )

    ratio_3d = (
        distance_3d(mcp, tip) /
        max(distance_3d(mcp, pip), 0.0001)
    )

    angle_ok = (
        angle_2d >= ANGLE_THRESHOLD_2D or
        angle_3d >= ANGLE_THRESHOLD_3D
    )

    length_ok = (
        ratio_2d >= LENGTH_RATIO_THRESHOLD or
        ratio_3d >= LENGTH_RATIO_THRESHOLD
    )

    tip_ok = (
        dip_angle_2d >= 110 or
        dip_angle_3d >= 110
    )

    return angle_ok and length_ok and tip_ok


# ============================================================
# 10. GESTURE STABILIZATION
# ============================================================

def update_gesture(raw_extended):

    global finger_up_count
    global finger_down_count
    global index_is_extended

    if raw_extended:

        finger_up_count += 1
        finger_down_count = 0

    else:

        finger_down_count += 1
        finger_up_count = 0

    if finger_up_count >= UP_THRESHOLD:
        index_is_extended = True

    elif finger_down_count >= DOWN_THRESHOLD:
        index_is_extended = False

    return index_is_extended


# ============================================================
# 11. MOVEMENT FILTER RESET
# ============================================================

def reset_movement_filter():

    global raw_point_history
    global stationary_anchor
    global filtered_point
    global movement_active

    raw_point_history = []

    stationary_anchor = None
    filtered_point = None
    movement_active = False


def reset_tracking():

    global previous_point
    global smoothed_point
    global finger_up_count
    global finger_down_count
    global index_is_extended

    previous_point = None
    smoothed_point = None

    finger_up_count = 0
    finger_down_count = 0

    index_is_extended = False

    reset_movement_filter()


# ============================================================
# 12. EMA CURSOR SMOOTHING
# ============================================================

def smooth_point(new_point, old_point):

    if old_point is None:
        return (
            float(new_point[0]),
            float(new_point[1])
        )

    x = (
        SMOOTHING * new_point[0] +
        (1 - SMOOTHING) * old_point[0]
    )

    y = (
        SMOOTHING * new_point[1] +
        (1 - SMOOTHING) * old_point[1]
    )

    return x, y


# ============================================================
# 13. STATIONARY FINGER FILTER
# ============================================================

def stabilize_finger(raw_point):

    global raw_point_history
    global stationary_anchor
    global filtered_point
    global movement_active

    raw_x, raw_y = raw_point

    # First detected position
    if stationary_anchor is None:

        stationary_anchor = (raw_x, raw_y)
        filtered_point = (float(raw_x), float(raw_y))

        raw_point_history = [(raw_x, raw_y)]

        return filtered_point, False

    # --------------------------------------------------------
    # STATIONARY MODE
    # --------------------------------------------------------

    if not movement_active:

        movement = point_distance(
            raw_point,
            stationary_anchor
        )

        # Ignore small landmark movements
        if movement < MOVEMENT_START_THRESHOLD:

            return filtered_point, False

        # Real movement detected
        movement_active = True

        raw_point_history = [(raw_x, raw_y)]

        filtered_point = smooth_point(
            raw_point,
            filtered_point
        )

        return filtered_point, True

    # --------------------------------------------------------
    # ACTIVE MOVEMENT MODE
    # --------------------------------------------------------

    # Keep recent raw coordinates
    raw_point_history.append((raw_x, raw_y))

    if len(raw_point_history) > STATIONARY_FRAMES:
        raw_point_history.pop(0)

    # Smooth the cursor
    filtered_point = smooth_point(
        raw_point,
        filtered_point
    )

    # --------------------------------------------------------
    # CHECK IF FINGER HAS STOPPED
    # --------------------------------------------------------

    if len(raw_point_history) >= STATIONARY_FRAMES:

        history = np.array(
            raw_point_history,
            dtype=np.float32
        )

        center = np.mean(history, axis=0)

        distances = np.linalg.norm(
            history - center,
            axis=1
        )

        spread = np.max(distances)

        # If recent positions are close together,
        # consider the finger stationary again.
        if spread <= STATIONARY_RADIUS:

            movement_active = False

            stationary_anchor = (raw_x, raw_y)

            raw_point_history = []

            # Keep the last stable cursor position
            return filtered_point, False

    return filtered_point, True


# ============================================================
# 14. CATMULL-ROM CURVE
# ============================================================

def catmull_rom(p0, p1, p2, p3):

    result = []

    p0 = np.array(p0, dtype=np.float64)
    p1 = np.array(p1, dtype=np.float64)
    p2 = np.array(p2, dtype=np.float64)
    p3 = np.array(p3, dtype=np.float64)

    for i in range(CURVE_STEPS):

        t = i / CURVE_STEPS
        t2 = t * t
        t3 = t2 * t

        point = 0.5 * (
            2 * p1
            + (-p0 + p2) * t
            + (2*p0 - 5*p1 + 4*p2 - p3) * t2
            + (-p0 + 3*p1 - 3*p2 + p3) * t3
        )

        result.append((
            int(round(point[0])),
            int(round(point[1]))
        ))

    return result


def get_smooth_curve(points):

    if len(points) <= 2:
        return [
            (int(round(p[0])), int(round(p[1])))
            for p in points
        ]

    result = []

    extended = [points[0]] + points + [points[-1]]

    for i in range(1, len(extended) - 2):

        result.extend(
            catmull_rom(
                extended[i - 1],
                extended[i],
                extended[i + 1],
                extended[i + 2]
            )
        )

    result.append((
        int(round(points[-1][0])),
        int(round(points[-1][1]))
    ))

    return result


# ============================================================
# 15. DRAWING FUNCTION
# ============================================================

def draw_smooth_stroke(image, points, color, tool="PEN"):

    if not points:
        return

    settings = TOOLS.get(tool, TOOLS["PEN"])

    thickness = settings["thickness"]
    opacity = settings["opacity"]

    radius = max(1, thickness // 2)

    smooth_points = get_smooth_curve(points)

    # Separate layer prevents brush blur from damaging old strokes
    layer = np.zeros_like(image)

    if len(smooth_points) == 1:

        cv2.circle(
            layer,
            smooth_points[0],
            radius,
            color,
            -1,
            cv2.LINE_AA
        )

    else:

        pts = np.array(
            smooth_points,
            dtype=np.int32
        ).reshape((-1, 1, 2))

        draw_color = color

        if tool == "PENCIL":

            draw_color = tuple(
                int(c * 0.85 + 255 * 0.15)
                for c in color
            )

        cv2.polylines(
            layer,
            [pts],
            False,
            draw_color,
            thickness,
            cv2.LINE_AA
        )

        cv2.circle(
            layer,
            smooth_points[0],
            radius,
            draw_color,
            -1,
            cv2.LINE_AA
        )

        cv2.circle(
            layer,
            smooth_points[-1],
            radius,
            draw_color,
            -1,
            cv2.LINE_AA
        )

    if tool == "PENCIL" and len(smooth_points) == 1:

        pencil_color = tuple(
            int(c * 0.85 + 255 * 0.15)
            for c in color
        )

        layer[:] = 0

        cv2.circle(
            layer,
            smooth_points[0],
            radius,
            pencil_color,
            -1,
            cv2.LINE_AA
        )

    elif tool == "BRUSH":

        layer = cv2.GaussianBlur(
            layer,
            (5, 5),
            1.0
        )

    mask = cv2.cvtColor(
        layer,
        cv2.COLOR_BGR2GRAY
    )

    active = mask > 0

    if not np.any(active):
        return

    if opacity >= 1.0:

        image[active] = layer[active]

    else:

        blended = (
            layer.astype(np.float32) * opacity +
            image.astype(np.float32) * (1 - opacity)
        )

        image[active] = np.clip(
            blended[active],
            0,
            255
        ).astype(np.uint8)


# ============================================================
# 16. SAVE STROKE
# ============================================================

def save_current_stroke():

    global current_stroke
    global strokes
    global undo_history

    if not current_stroke:
        return

    stroke = {
        "points": current_stroke.copy(),
        "color": current_stroke_color,
        "tool": current_stroke_tool
    }

    strokes.append(stroke)

    undo_history.append(stroke)
    undo_history = undo_history[-MAX_UNDO:]

    draw_smooth_stroke(
        canvas,
        stroke["points"],
        stroke["color"],
        stroke["tool"]
    )

    current_stroke = []


# ============================================================
# 17. REBUILD CANVAS
# ============================================================

def rebuild_canvas():

    canvas[:] = 255

    for stroke in strokes:

        draw_smooth_stroke(
            canvas,
            stroke["points"],
            stroke["color"],
            stroke["tool"]
        )


# ============================================================
# 18. COLOR PALETTE
# ============================================================

def draw_color_palette(image, selected_index):

    start_x = 20
    start_y = 23

    spacing = 32
    radius = 6

    cv2.rectangle(
        image,
        (8, 7),
        (start_x + 7 * spacing + 12, 48),
        (245, 245, 245),
        -1
    )

    for i, (name, color) in enumerate(COLORS):

        x = start_x + i * spacing
        y = start_y

        cv2.circle(
            image,
            (x, y),
            radius,
            color,
            -1,
            cv2.LINE_AA
        )

        if i == selected_index:

            cv2.circle(
                image,
                (x, y),
                radius + 3,
                (0, 0, 0),
                1,
                cv2.LINE_AA
            )

        cv2.putText(
            image,
            str(i + 1),
            (x - 3, y + 17),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.3,
            (0, 0, 0),
            1,
            cv2.LINE_AA
        )


# ============================================================
# 19. TOOL STATUS
# ============================================================

def draw_tool_status(image):

    text = (
        f"TOOL: {selected_tool} | "
        f"SIZE: {TOOLS[selected_tool]['thickness']}"
    )

    cv2.putText(
        image,
        text,
        (20, 135),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.55,
        (40, 40, 40),
        2,
        cv2.LINE_AA
    )


# ============================================================
# 20. MAIN APPLICATION
# ============================================================

try:

    with mp_hands.Hands(

        static_image_mode=False,

        max_num_hands=1,

        model_complexity=1,

        min_detection_confidence=MIN_DETECTION_CONFIDENCE,

        min_tracking_confidence=MIN_TRACKING_CONFIDENCE

    ) as hands:

        while True:

            success, frame = cap.read()

            if not success:
                print("Unable to read camera frame.")
                break

            frame = cv2.flip(frame, 1)

            height, width = frame.shape[:2]

            if canvas is None:
                canvas = np.full_like(frame, 255)

            # ----------------------------------------------
            # CONTINUOUS HAND DETECTION
            # ----------------------------------------------

            small_frame = cv2.resize(
                frame,
                (PROCESS_WIDTH, PROCESS_HEIGHT),
                interpolation=cv2.INTER_AREA
            )

            rgb = cv2.cvtColor(
                small_frame,
                cv2.COLOR_BGR2RGB
            )

            rgb.flags.writeable = False
            results = hands.process(rgb)
            rgb.flags.writeable = True

            hand_detected = bool(results.multi_hand_landmarks)

            status = "NO HAND DETECTED - SHOW YOUR HAND"
            status_color = (0, 0, 255)

            cursor_point = None

            debug_frame = small_frame.copy()

            if hand_detected:

                hand = results.multi_hand_landmarks[0]

                mp_drawing.draw_landmarks(
                    debug_frame,
                    hand,
                    mp_hands.HAND_CONNECTIONS
                )

                tip = hand.landmark[8]

                x = int(tip.x * width)
                y = int(tip.y * height)

                x = max(0, min(width - 1, x))
                y = max(0, min(height - 1, y))

                lost_hand_frames = 0

                raw_extended = is_index_finger_extended(hand)

                index_is_extended = update_gesture(raw_extended)

                if not drawing_started:

                    status = "HAND DETECTED - PRESS S TO START"
                    status_color = (0, 150, 0)

                elif drawing_paused:

                    status = "PAUSED - PRESS P TO RESUME"
                    status_color = (0, 0, 255)

                    save_current_stroke()
                    reset_tracking()

                elif index_is_extended:

                    status = "DRAWING"
                    status_color = brush_color

                    # Apply stationary finger filter
                    stable_point, actual_movement = stabilize_finger(
                        (x, y)
                    )

                    # Apply cursor smoothing
                    smoothed_point = stable_point

                    cursor_point = (
                        int(round(smoothed_point[0])),
                        int(round(smoothed_point[1]))
                    )

                    if previous_point is None:

                        # Start a new stroke
                        current_stroke = [cursor_point]

                        current_stroke_color = brush_color
                        current_stroke_tool = selected_tool

                        previous_point = smoothed_point

                    elif actual_movement:

                        movement = point_distance(
                            smoothed_point,
                            previous_point
                        )

                        if movement >= MIN_POINT_DISTANCE:

                            current_stroke.append(cursor_point)

                            previous_point = smoothed_point

                else:

                    status = "HAND DETECTED - EXTEND INDEX FINGER"
                    status_color = (0, 140, 255)

                    save_current_stroke()

                    previous_point = None
                    smoothed_point = None

                    reset_movement_filter()

            else:

                lost_hand_frames += 1

                if lost_hand_frames >= MAX_LOST_HAND_FRAMES:

                    save_current_stroke()
                    reset_tracking()

            # ----------------------------------------------
            # DEBUG WINDOW
            # ----------------------------------------------

            debug_text = (
                "HAND FOUND" if hand_detected else "NO HAND"
            )

            debug_color = (
                (0, 255, 0) if hand_detected else (0, 0, 255)
            )

            cv2.putText(
                debug_frame,
                debug_text,
                (10, 25),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                debug_color,
                2,
                cv2.LINE_AA
            )

            if SHOW_DEBUG_WINDOW:
                cv2.imshow("Hand Tracking Debug", debug_frame)

            # ----------------------------------------------
            # CANVAS DISPLAY
            # ----------------------------------------------

            output = canvas.copy()

            if current_stroke:

                draw_smooth_stroke(
                    output,
                    current_stroke,
                    current_stroke_color,
                    current_stroke_tool
                )

            draw_color_palette(
                output,
                selected_color_index
            )

            draw_tool_status(output)

            if (
                cursor_point is not None
                and drawing_started
                and not drawing_paused
            ):

                cv2.circle(
                    output,
                    cursor_point,
                    max(3, TOOLS[selected_tool]["thickness"] // 2),
                    brush_color,
                    -1,
                    cv2.LINE_AA
                )

            cv2.putText(
                output,
                status,
                (20, 100),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                status_color,
                2,
                cv2.LINE_AA
            )

            instructions = (
                "S: Start | 1-8: Color | N: Pen | M: Marker | "
                "X: Max | L: Pencil | B: Brush | "
                "P: Pause | Z: Undo | C: Clear | Q: Quit"
            )

            cv2.putText(
                output,
                instructions,
                (20, height - 20),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.43,
                (80, 80, 80),
                1,
                cv2.LINE_AA
            )

            cv2.imshow("Virtual Paint App", output)

            # ----------------------------------------------
            # KEYBOARD CONTROLS
            # ----------------------------------------------

            key = cv2.waitKey(1) & 0xFF

            if ord("1") <= key <= ord("8"):

                selected_color_index = key - ord("1")
                brush_color = COLORS[selected_color_index][1]

            elif key == ord("n"):
                selected_tool = "PEN"

            elif key == ord("m"):
                selected_tool = "MARKER"

            elif key == ord("x"):
                selected_tool = "MAX THICKNESS"

            elif key == ord("l"):
                selected_tool = "PENCIL"

            elif key == ord("b"):
                selected_tool = "BRUSH"

            elif key == ord("s") and not drawing_started:

                drawing_started = True
                drawing_paused = False

                reset_tracking()
                save_current_stroke()

            elif key == ord("p") and drawing_started:

                drawing_paused = not drawing_paused

                save_current_stroke()
                reset_tracking()

            elif key == ord("z") and drawing_started:

                save_current_stroke()

                if undo_history:

                    last_stroke = undo_history.pop()

                    for i in range(len(strokes) - 1, -1, -1):

                        if strokes[i] is last_stroke:
                            strokes.pop(i)
                            break

                    rebuild_canvas()

                reset_tracking()

            elif key == ord("c"):

                strokes.clear()
                current_stroke.clear()
                undo_history.clear()

                canvas[:] = 255

                drawing_started = False
                drawing_paused = False

                lost_hand_frames = 0
                reset_tracking()

            elif key == ord("q"):

                break

finally:

    cap.release()
    cv2.destroyAllWindows()
```

```python
import cv2
import mediapipe as mp
import numpy as np
import math
from collections import deque

# ============================================================
# 1. CAMERA SETUP
# ============================================================

CAMERA_INDEX = 0
FRAME_WIDTH = 1280
FRAME_HEIGHT = 720

cap = cv2.VideoCapture(CAMERA_INDEX)

if not cap.isOpened():
    cap.release()
    cap = cv2.VideoCapture(1)

if not cap.isOpened():
    raise RuntimeError("Camera not found. Check your camera connection.")

cap.set(cv2.CAP_PROP_FRAME_WIDTH, FRAME_WIDTH)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, FRAME_HEIGHT)
cap.set(cv2.CAP_PROP_FPS, 30)
cap.set(cv2.CAP_PROP_BUFFERSIZE, 1)

mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils

# ============================================================
# 2. COLORS
# ============================================================

COLORS = [
    ("Green",  (0, 200, 0)),
    ("Red",    (0, 0, 255)),
    ("Blue",   (255, 0, 0)),
    ("Black",  (0, 0, 0)),
    ("Yellow", (0, 255, 255)),
    ("Purple", (180, 0, 180)),
    ("Orange", (0, 140, 255)),
    ("Pink",   (203, 192, 255))
]

selected_color_index = 0
brush_color = COLORS[selected_color_index][1]

# ============================================================
# 3. DRAWING TOOLS
# ============================================================

TOOLS = {
    "PEN": {"thickness": 2, "opacity": 1.0},
    "MARKER": {"thickness": 9, "opacity": 0.55},
    "MAX THICKNESS": {"thickness": 24, "opacity": 1.0},
    "PENCIL": {"thickness": 2, "opacity": 0.80},
    "BRUSH": {"thickness": 15, "opacity": 0.90}
}

selected_tool = "PEN"
current_stroke_tool = selected_tool
current_stroke_color = brush_color

# ============================================================
# 4. CANVAS AND APP VARIABLES
# ============================================================

canvas = None

strokes = []
current_stroke = []
undo_history = []

MAX_UNDO = 10

drawing_started = False
drawing_paused = False

previous_point = None
smoothed_point = None

# ============================================================
# 5. HAND DETECTION SETTINGS
# ============================================================

PROCESS_WIDTH = 640
PROCESS_HEIGHT = 360

MIN_DETECTION_CONFIDENCE = 0.50
MIN_TRACKING_CONFIDENCE = 0.50

UP_THRESHOLD = 2
DOWN_THRESHOLD = 3

finger_up_count = 0
finger_down_count = 0
index_is_extended = False

lost_hand_frames = 0
MAX_LOST_HAND_FRAMES = 2

# ============================================================
# 6. ADVANCED DRAWING SETTINGS
# ============================================================

# Adaptive smoothing
MIN_SMOOTHING = 0.20
MAX_SMOOTHING = 0.55

# Minimum distance before adding a new drawing point
MIN_POINT_DISTANCE = 1.2

# Curve quality
CURVE_STEPS = 12

CURSOR_RADIUS = 6

# Stationary finger protection
MOVEMENT_START_THRESHOLD = 4.0
STATIONARY_RADIUS = 2.2
STATIONARY_FRAMES = 8

# Noise reduction
RAW_HISTORY_SIZE = 3
NOISE_DEAD_ZONE = 0.7

raw_point_history = deque(maxlen=RAW_HISTORY_SIZE)
stationary_history = deque(maxlen=STATIONARY_FRAMES)

stationary_anchor = None
filtered_point = None
movement_active = False

SHOW_DEBUG_WINDOW = True

# Finger detection
ANGLE_THRESHOLD_2D = 125
ANGLE_THRESHOLD_3D = 130
LENGTH_RATIO_THRESHOLD = 1.00

# ============================================================
# 7. DISTANCE FUNCTIONS
# ============================================================

def distance_2d(a, b):
    return math.hypot(a.x - b.x, a.y - b.y)


def distance_3d(a, b):
    return math.sqrt(
        (a.x - b.x) ** 2 +
        (a.y - b.y) ** 2 +
        (a.z - b.z) ** 2
    )


def point_distance(p1, p2):
    return math.hypot(
        p1[0] - p2[0],
        p1[1] - p2[1]
    )


# ============================================================
# 8. ANGLE FUNCTIONS
# ============================================================

def calculate_angle_2d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


def calculate_angle_3d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y, a.z - b.z],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y, c.z - b.z],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


# ============================================================
# 9. INDEX FINGER DETECTION
# ============================================================

def is_index_finger_extended(hand):

    lm = hand.landmark

    mcp = lm[5]
    pip = lm[6]
    dip = lm[7]
    tip = lm[8]

    angle_2d = calculate_angle_2d(mcp, pip, tip)
    angle_3d = calculate_angle_3d(mcp, pip, tip)

    dip_angle_2d = calculate_angle_2d(pip, dip, tip)
    dip_angle_3d = calculate_angle_3d(pip, dip, tip)

    ratio_2d = (
        distance_2d(mcp, tip) /
        max(distance_2d(mcp, pip), 0.0001)
    )

    ratio_3d = (
        distance_3d(mcp, tip) /
        max(distance_3d(mcp, pip), 0.0001)
    )

    angle_ok = (
        angle_2d >= ANGLE_THRESHOLD_2D or
        angle_3d >= ANGLE_THRESHOLD_3D
    )

    length_ok = (
        ratio_2d >= LENGTH_RATIO_THRESHOLD or
        ratio_3d >= LENGTH_RATIO_THRESHOLD
    )

    tip_ok = (
        dip_angle_2d >= 110 or
        dip_angle_3d >= 110
    )

    return angle_ok and length_ok and tip_ok


# ============================================================
# 10. GESTURE STABILIZATION
# ============================================================

def update_gesture(raw_extended):

    global finger_up_count
    global finger_down_count
    global index_is_extended

    if raw_extended:

        finger_up_count += 1
        finger_down_count = 0

    else:

        finger_down_count += 1
        finger_up_count = 0

    if finger_up_count >= UP_THRESHOLD:
        index_is_extended = True

    elif finger_down_count >= DOWN_THRESHOLD:
        index_is_extended = False

    return index_is_extended


# ============================================================
# 11. RESET DRAWING FILTER
# ============================================================

def reset_movement_filter():

    global raw_point_history
    global stationary_history
    global stationary_anchor
    global filtered_point
    global movement_active

    raw_point_history.clear()
    stationary_history.clear()

    stationary_anchor = None
    filtered_point = None
    movement_active = False


def reset_tracking():

    global previous_point
    global smoothed_point
    global finger_up_count
    global finger_down_count
    global index_is_extended

    previous_point = None
    smoothed_point = None

    finger_up_count = 0
    finger_down_count = 0

    index_is_extended = False

    reset_movement_filter()


# ============================================================
# 12. RAW POINT NOISE REDUCTION
# ============================================================

def get_median_point(raw_point):

    raw_point_history.append(raw_point)

    points = np.array(
        list(raw_point_history),
        dtype=np.float32
    )

    median_x = float(np.median(points[:, 0]))
    median_y = float(np.median(points[:, 1]))

    return median_x, median_y


# ============================================================
# 13. ADAPTIVE SMOOTHING
# ============================================================

def adaptive_smooth(new_point, old_point):

    if old_point is None:

        return (
            float(new_point[0]),
            float(new_point[1])
        )

    movement = point_distance(
        new_point,
        old_point
    )

    # Slow movements receive more smoothing.
    # Fast movements receive less smoothing delay.
    speed_factor = min(movement / 18.0, 1.0)

    alpha = (
        MIN_SMOOTHING +
        (MAX_SMOOTHING - MIN_SMOOTHING) * speed_factor
    )

    dx = new_point[0] - old_point[0]
    dy = new_point[1] - old_point[1]

    return (
        old_point[0] + alpha * dx,
        old_point[1] + alpha * dy
    )


# ============================================================
# 14. STATIONARY AND MOVEMENT FILTER
# ============================================================

def stabilize_finger(raw_point):

    global stationary_anchor
    global filtered_point
    global movement_active
    global stationary_history

    # First remove small tracking spikes
    clean_point = get_median_point(raw_point)

    # First detected position
    if stationary_anchor is None:

        stationary_anchor = clean_point

        filtered_point = clean_point

        stationary_history.clear()
        stationary_history.append(clean_point)

        return clean_point, False, False

    # --------------------------------------------------------
    # STATIONARY MODE
    # --------------------------------------------------------

    if not movement_active:

        movement = point_distance(
            clean_point,
            stationary_anchor
        )

        # Ignore tiny finger movements
        if movement < MOVEMENT_START_THRESHOLD:

            return filtered_point, False, False

        # Actual movement has started
        movement_active = True

        stationary_history.clear()

        filtered_point = clean_point

        return filtered_point, True, True

    # --------------------------------------------------------
    # ACTIVE MOVEMENT MODE
    # --------------------------------------------------------

    # Apply adaptive smoothing
    filtered_point = adaptive_smooth(
        clean_point,
        filtered_point
    )

    stationary_history.append(clean_point)

    # --------------------------------------------------------
    # CHECK WHETHER FINGER HAS STOPPED
    # --------------------------------------------------------

    if len(stationary_history) >= STATIONARY_FRAMES:

        history = np.array(
            list(stationary_history),
            dtype=np.float32
        )

        center = np.median(
            history,
            axis=0
        )

        distances = np.linalg.norm(
            history - center,
            axis=1
        )

        spread = float(np.max(distances))

        if spread <= STATIONARY_RADIUS:

            movement_active = False

            stationary_anchor = clean_point

            stationary_history.clear()

            return filtered_point, False, False

    return filtered_point, True, False


# ============================================================
# 15. CLEAN DRAWING POINTS
# ============================================================

def add_drawing_point(points, new_point):

    if not points:

        points.append(new_point)
        return

    last_point = points[-1]

    distance = point_distance(
        last_point,
        new_point
    )

    # Ignore tiny changes, but let slow movement accumulate.
    if distance >= MIN_POINT_DISTANCE:

        points.append(new_point)


# ============================================================
# 16. CATMULL-ROM CURVE
# ============================================================

def catmull_rom(p0, p1, p2, p3):

    result = []

    p0 = np.array(p0, dtype=np.float64)
    p1 = np.array(p1, dtype=np.float64)
    p2 = np.array(p2, dtype=np.float64)
    p3 = np.array(p3, dtype=np.float64)

    for i in range(CURVE_STEPS):

        t = i / CURVE_STEPS

        t2 = t * t
        t3 = t2 * t

        point = 0.5 * (
            2 * p1
            + (-p0 + p2) * t
            + (2*p0 - 5*p1 + 4*p2 - p3) * t2
            + (-p0 + 3*p1 - 3*p2 + p3) * t3
        )

        result.append((
            float(point[0]),
            float(point[1])
        ))

    return result


def get_smooth_curve(points):

    if len(points) <= 2:
        return [
            (
                int(round(p[0])),
                int(round(p[1]))
            )
            for p in points
        ]

    result = []

    extended = (
        [points[0]] +
        points +
        [points[-1]]
    )

    for i in range(1, len(extended) - 2):

        result.extend(
            catmull_rom(
                extended[i - 1],
                extended[i],
                extended[i + 1],
                extended[i + 2]
            )
        )

    result.append((
        int(round(points[-1][0])),
        int(round(points[-1][1]))
    ))

    return [
        (
            int(round(p[0])),
            int(round(p[1]))
        )
        for p in result
    ]


# ============================================================
# 17. DRAWING FUNCTION
# ============================================================

def draw_smooth_stroke(image, points, color, tool="PEN"):

    if not points:
        return

    settings = TOOLS.get(tool, TOOLS["PEN"])

    thickness = settings["thickness"]
    opacity = settings["opacity"]

    radius = max(1, thickness // 2)

    smooth_points = get_smooth_curve(points)

    layer = np.zeros_like(image)

    draw_color = color

    if tool == "PENCIL":

        draw_color = tuple(
            int(c * 0.85 + 255 * 0.15)
            for c in color
        )

    if len(smooth_points) == 1:

        cv2.circle(
            layer,
            smooth_points[0],
            radius,
            draw_color,
            -1,
            cv2.LINE_AA
        )

    else:

        pts = np.array(
            smooth_points,
            dtype=np.int32
        ).reshape((-1, 1, 2))

        cv2.polylines(
            layer,
            [pts],
            False,
            draw_color,
            thickness,
            cv2.LINE_AA
        )

        # Rounded beginning and ending
        cv2.circle(
            layer,
            smooth_points[0],
            radius,
            draw_color,
            -1,
            cv2.LINE_AA
        )

        cv2.circle(
            layer,
            smooth_points[-1],
            radius,
            draw_color,
            -1,
            cv2.LINE_AA
        )

    # Brush softness
    if tool == "BRUSH":

        layer = cv2.GaussianBlur(
            layer,
            (5, 5),
            1.0
        )

    mask = cv2.cvtColor(
        layer,
        cv2.COLOR_BGR2GRAY
    )

    active = mask > 0

    if not np.any(active):
        return

    if opacity >= 1.0:

        image[active] = layer[active]

    else:

        blended = (
            layer.astype(np.float32) * opacity +
            image.astype(np.float32) * (1 - opacity)
        )

        image[active] = np.clip(
            blended[active],
            0,
            255
        ).astype(np.uint8)


# ============================================================
# 18. SAVE STROKE
# ============================================================

def save_current_stroke():

    global current_stroke
    global strokes
    global undo_history

    if not current_stroke:
        return

    stroke = {
        "points": current_stroke.copy(),
        "color": current_stroke_color,
        "tool": current_stroke_tool
    }

    strokes.append(stroke)

    undo_history.append(stroke)
    undo_history = undo_history[-MAX_UNDO:]

    draw_smooth_stroke(
        canvas,
        stroke["points"],
        stroke["color"],
        stroke["tool"]
    )

    current_stroke = []


# ============================================================
# 19. REBUILD CANVAS
# ============================================================

def rebuild_canvas():

    canvas[:] = 255

    for stroke in strokes:

        draw_smooth_stroke(
            canvas,
            stroke["points"],
            stroke["color"],
            stroke["tool"]
        )


# ============================================================
# 20. COLOR PALETTE
# ============================================================

def draw_color_palette(image, selected_index):

    start_x = 20
    start_y = 23

    spacing = 32
    radius = 6

    cv2.rectangle(
        image,
        (8, 7),
        (start_x + 7 * spacing + 12, 48),
        (245, 245, 245),
        -1
    )

    for i, (name, color) in enumerate(COLORS):

        x = start_x + i * spacing
        y = start_y

        cv2.circle(
            image,
            (x, y),
            radius,
            color,
            -1,
            cv2.LINE_AA
        )

        if i == selected_index:

            cv2.circle(
                image,
                (x, y),
                radius + 3,
                (0, 0, 0),
                1,
                cv2.LINE_AA
            )

        cv2.putText(
            image,
            str(i + 1),
            (x - 3, y + 17),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.3,
            (0, 0, 0),
            1,
            cv2.LINE_AA
        )


# ============================================================
# 21. TOOL STATUS
# ============================================================

def draw_tool_status(image):

    text = (
        f"TOOL: {selected_tool} | "
        f"SIZE: {TOOLS[selected_tool]['thickness']}"
    )

    cv2.putText(
        image,
        text,
        (20, 135),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.55,
        (40, 40, 40),
        2,
        cv2.LINE_AA
    )


# ============================================================
# 22. MAIN APPLICATION
# ============================================================

try:

    with mp_hands.Hands(

        static_image_mode=False,

        max_num_hands=1,

        model_complexity=1,

        min_detection_confidence=MIN_DETECTION_CONFIDENCE,

        min_tracking_confidence=MIN_TRACKING_CONFIDENCE

    ) as hands:

        while True:

            success, frame = cap.read()

            if not success:
                print("Unable to read camera frame.")
                break

            frame = cv2.flip(frame, 1)

            height, width = frame.shape[:2]

            if canvas is None:
                canvas = np.full_like(frame, 255)

            # ----------------------------------------------
            # HAND DETECTION
            # ----------------------------------------------

            small_frame = cv2.resize(
                frame,
                (PROCESS_WIDTH, PROCESS_HEIGHT),
                interpolation=cv2.INTER_AREA
            )

            rgb = cv2.cvtColor(
                small_frame,
                cv2.COLOR_BGR2RGB
            )

            rgb.flags.writeable = False
            results = hands.process(rgb)
            rgb.flags.writeable = True

            hand_detected = bool(results.multi_hand_landmarks)

            status = "NO HAND DETECTED - SHOW YOUR HAND"
            status_color = (0, 0, 255)

            cursor_point = None

            debug_frame = small_frame.copy()

            if hand_detected:

                hand = results.multi_hand_landmarks[0]

                mp_drawing.draw_landmarks(
                    debug_frame,
                    hand,
                    mp_hands.HAND_CONNECTIONS
                )

                tip = hand.landmark[8]

                x = int(tip.x * width)
                y = int(tip.y * height)

                x = max(0, min(width - 1, x))
                y = max(0, min(height - 1, y))

                lost_hand_frames = 0

                raw_extended = is_index_finger_extended(hand)

                index_is_extended = update_gesture(raw_extended)

                if not drawing_started:

                    status = "HAND DETECTED - PRESS S TO START"
                    status_color = (0, 150, 0)

                elif drawing_paused:

                    status = "PAUSED - PRESS P TO RESUME"
                    status_color = (0, 0, 255)

                    save_current_stroke()
                    reset_tracking()

                elif index_is_extended:

                    status = "DRAWING"
                    status_color = brush_color

                    stable_point, actual_movement, movement_started = (
                        stabilize_finger((x, y))
                    )

                    smoothed_point = stable_point

                    cursor_point = (
                        int(round(stable_point[0])),
                        int(round(stable_point[1]))
                    )

                    # Start stroke only when real movement begins.
                    if movement_started:

                        current_stroke = []

                        current_stroke_color = brush_color
                        current_stroke_tool = selected_tool

                        start_point = (
                            int(round(stationary_anchor[0])),
                            int(round(stationary_anchor[1]))
                        )

                        current_stroke.append(start_point)

                        previous_point = (
                            float(start_point[0]),
                            float(start_point[1])
                        )

                    if actual_movement and previous_point is not None:

                        movement = point_distance(
                            stable_point,
                            previous_point
                        )

                        if movement >= MIN_POINT_DISTANCE:

                            add_drawing_point(
                                current_stroke,
                                cursor_point
                            )

                            previous_point = stable_point

                else:

                    status = "HAND DETECTED - EXTEND INDEX FINGER"
                    status_color = (0, 140, 255)

                    save_current_stroke()

                    previous_point = None
                    smoothed_point = None

                    reset_movement_filter()

            else:

                lost_hand_frames += 1

                if lost_hand_frames >= MAX_LOST_HAND_FRAMES:

                    save_current_stroke()
                    reset_tracking()

            # ----------------------------------------------
            # DEBUG WINDOW
            # ----------------------------------------------

            debug_text = (
                "HAND FOUND" if hand_detected else "NO HAND"
            )

            debug_color = (
                (0, 255, 0) if hand_detected else (0, 0, 255)
            )

            cv2.putText(
                debug_frame,
                debug_text,
                (10, 25),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                debug_color,
                2,
                cv2.LINE_AA
            )

            if SHOW_DEBUG_WINDOW:
                cv2.imshow("Hand Tracking Debug", debug_frame)

            # ----------------------------------------------
            # CANVAS DISPLAY
            # ----------------------------------------------

            output = canvas.copy()

            if current_stroke:

                draw_smooth_stroke(
                    output,
                    current_stroke,
                    current_stroke_color,
                    current_stroke_tool
                )

            draw_color_palette(
                output,
                selected_color_index
            )

            draw_tool_status(output)

            if (
                cursor_point is not None
                and drawing_started
                and not drawing_paused
            ):

                cv2.circle(
                    output,
                    cursor_point,
                    max(3, TOOLS[selected_tool]["thickness"] // 2),
                    brush_color,
                    -1,
                    cv2.LINE_AA
                )

            cv2.putText(
                output,
                status,
                (20, 100),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                status_color,
                2,
                cv2.LINE_AA
            )

            instructions = (
                "S: Start | 1-8: Color | N: Pen | M: Marker | "
                "X: Max | L: Pencil | B: Brush | "
                "P: Pause | Z: Undo | C: Clear | Q: Quit"
            )

            cv2.putText(
                output,
                instructions,
                (20, height - 20),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.43,
                (80, 80, 80),
                1,
                cv2.LINE_AA
            )

            cv2.imshow("Virtual Paint App", output)

            # ----------------------------------------------
            # KEYBOARD CONTROLS
            # ----------------------------------------------

            key = cv2.waitKey(1) & 0xFF

            if ord("1") <= key <= ord("8"):

                selected_color_index = key - ord("1")
                brush_color = COLORS[selected_color_index][1]

            elif key == ord("n"):
                selected_tool = "PEN"

            elif key == ord("m"):
                selected_tool = "MARKER"

            elif key == ord("x"):
                selected_tool = "MAX THICKNESS"

            elif key == ord("l"):
                selected_tool = "PENCIL"

            elif key == ord("b"):
                selected_tool = "BRUSH"

            elif key == ord("s") and not drawing_started:

                drawing_started = True
                drawing_paused = False

                reset_tracking()
                save_current_stroke()

            elif key == ord("p") and drawing_started:

                drawing_paused = not drawing_paused

                save_current_stroke()
                reset_tracking()

            elif key == ord("z") and drawing_started:

                save_current_stroke()

                if undo_history:

                    last_stroke = undo_history.pop()

                    for i in range(len(strokes) - 1, -1, -1):

                        if strokes[i] is last_stroke:
                            strokes.pop(i)
                            break

                    rebuild_canvas()

                reset_tracking()

            elif key == ord("c"):

                strokes.clear()
                current_stroke.clear()
                undo_history.clear()

                canvas[:] = 255

                drawing_started = False
                drawing_paused = False

                lost_hand_frames = 0
                reset_tracking()

            elif key == ord("q"):

                break

finally:

    cap.release()
    cv2.destroyAllWindows()
```

```python

import cv2
import mediapipe as mp
import numpy as np
import math
from collections import deque

# ============================================================
# 1. CAMERA SETUP
# ============================================================

CAMERA_INDEX = 0
FRAME_WIDTH = 1280
FRAME_HEIGHT = 720

cap = cv2.VideoCapture(CAMERA_INDEX)

if not cap.isOpened():
    cap.release()
    cap = cv2.VideoCapture(1)

if not cap.isOpened():
    raise RuntimeError("Camera not found. Check your camera connection.")

cap.set(cv2.CAP_PROP_FRAME_WIDTH, FRAME_WIDTH)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, FRAME_HEIGHT)
cap.set(cv2.CAP_PROP_FPS, 30)
cap.set(cv2.CAP_PROP_BUFFERSIZE, 1)

mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils

# ============================================================
# 2. COLORS
# ============================================================

COLORS = [
    ("Green",  (0, 200, 0)),
    ("Red",    (0, 0, 255)),
    ("Blue",   (255, 0, 0)),
    ("Black",  (0, 0, 0)),
    ("Yellow", (0, 255, 255)),
    ("Purple", (180, 0, 180)),
    ("Orange", (0, 140, 255)),
    ("Pink",   (203, 192, 255))
]

selected_color_index = 0
brush_color = COLORS[selected_color_index][1]

# ============================================================
# 3. DRAWING TOOLS
# ============================================================

TOOLS = {
    "PEN": {"thickness": 2, "opacity": 1.0},
    "MARKER": {"thickness": 9, "opacity": 0.55},
    "MAX THICKNESS": {"thickness": 24, "opacity": 1.0},
    "PENCIL": {"thickness": 2, "opacity": 0.80},
    "BRUSH": {"thickness": 15, "opacity": 0.90}
}

selected_tool = "PEN"
current_stroke_tool = selected_tool
current_stroke_color = brush_color

# ============================================================
# 4. CANVAS AND APP VARIABLES
# ============================================================

canvas = None

strokes = []
current_stroke = []
undo_history = []

MAX_UNDO = 10

drawing_started = False
drawing_paused = False

previous_point = None
smoothed_point = None

# ============================================================
# 5. HAND DETECTION SETTINGS
# ============================================================

PROCESS_WIDTH = 640
PROCESS_HEIGHT = 360

MIN_DETECTION_CONFIDENCE = 0.50
MIN_TRACKING_CONFIDENCE = 0.50

UP_THRESHOLD = 2
DOWN_THRESHOLD = 3

finger_up_count = 0
finger_down_count = 0
index_is_extended = False

lost_hand_frames = 0
MAX_LOST_HAND_FRAMES = 2

# ============================================================
# 6. ADVANCED DRAWING SETTINGS
# ============================================================

# Adaptive smoothing
MIN_SMOOTHING = 0.24
MAX_SMOOTHING = 0.62

# Minimum drawing point distance
MIN_POINT_DISTANCE = 1.0

# Curve quality
CURVE_STEPS = 14

CURSOR_RADIUS = 6

# Stationary finger protection
MOVEMENT_START_THRESHOLD = 3.0
STATIONARY_RADIUS = 1.8
STATIONARY_FRAMES = 7

# Noise reduction
RAW_HISTORY_SIZE = 3
NOISE_DEAD_ZONE = 0.7

# Edge stability
EDGE_MARGIN = 0.035

raw_point_history = deque(maxlen=RAW_HISTORY_SIZE)
stationary_history = deque(maxlen=STATIONARY_FRAMES)

stationary_anchor = None
filtered_point = None
movement_active = False

SHOW_DEBUG_WINDOW = True

# Finger detection
ANGLE_THRESHOLD_2D = 125
ANGLE_THRESHOLD_3D = 130
LENGTH_RATIO_THRESHOLD = 1.00

# ============================================================
# 7. DISTANCE FUNCTIONS
# ============================================================

def distance_2d(a, b):

    return math.hypot(
        a.x - b.x,
        a.y - b.y
    )


def distance_3d(a, b):

    return math.sqrt(
        (a.x - b.x) ** 2 +
        (a.y - b.y) ** 2 +
        (a.z - b.z) ** 2
    )


def point_distance(p1, p2):

    return math.hypot(
        p1[0] - p2[0],
        p1[1] - p2[1]
    )


# ============================================================
# 8. ANGLE FUNCTIONS
# ============================================================

def calculate_angle_2d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


def calculate_angle_3d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y, a.z - b.z],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y, c.z - b.z],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


# ============================================================
# 9. INDEX FINGER DETECTION
# ============================================================

def is_index_finger_extended(hand):

    lm = hand.landmark

    mcp = lm[5]
    pip = lm[6]
    dip = lm[7]
    tip = lm[8]

    angle_2d = calculate_angle_2d(mcp, pip, tip)
    angle_3d = calculate_angle_3d(mcp, pip, tip)

    dip_angle_2d = calculate_angle_2d(pip, dip, tip)
    dip_angle_3d = calculate_angle_3d(pip, dip, tip)

    ratio_2d = (
        distance_2d(mcp, tip) /
        max(distance_2d(mcp, pip), 0.0001)
    )

    ratio_3d = (
        distance_3d(mcp, tip) /
        max(distance_3d(mcp, pip), 0.0001)
    )

    angle_ok = (
        angle_2d >= ANGLE_THRESHOLD_2D or
        angle_3d >= ANGLE_THRESHOLD_3D
    )

    length_ok = (
        ratio_2d >= LENGTH_RATIO_THRESHOLD or
        ratio_3d >= LENGTH_RATIO_THRESHOLD
    )

    tip_ok = (
        dip_angle_2d >= 110 or
        dip_angle_3d >= 110
    )

    return angle_ok and length_ok and tip_ok


# ============================================================
# 10. GESTURE STABILIZATION
# ============================================================

def update_gesture(raw_extended):

    global finger_up_count
    global finger_down_count
    global index_is_extended

    if raw_extended:

        finger_up_count += 1
        finger_down_count = 0

    else:

        finger_down_count += 1
        finger_up_count = 0

    if finger_up_count >= UP_THRESHOLD:
        index_is_extended = True

    elif finger_down_count >= DOWN_THRESHOLD:
        index_is_extended = False

    return index_is_extended


# ============================================================
# 11. RESET DRAWING FILTER
# ============================================================

def reset_movement_filter():

    global stationary_anchor
    global filtered_point
    global movement_active

    raw_point_history.clear()
    stationary_history.clear()

    stationary_anchor = None
    filtered_point = None
    movement_active = False


def reset_tracking():

    global previous_point
    global smoothed_point
    global finger_up_count
    global finger_down_count
    global index_is_extended

    previous_point = None
    smoothed_point = None

    finger_up_count = 0
    finger_down_count = 0

    index_is_extended = False

    reset_movement_filter()


# ============================================================
# 12. RAW POINT NOISE REDUCTION
# ============================================================

def get_median_point(raw_point):

    raw_point_history.append(raw_point)

    points = np.array(
        list(raw_point_history),
        dtype=np.float32
    )

    median_x = float(np.median(points[:, 0]))
    median_y = float(np.median(points[:, 1]))

    return median_x, median_y


# ============================================================
# 13. ADAPTIVE SMOOTHING
# ============================================================

def adaptive_smooth(new_point, old_point):

    if old_point is None:

        return (
            float(new_point[0]),
            float(new_point[1])
        )

    dx = new_point[0] - old_point[0]
    dy = new_point[1] - old_point[1]

    movement = math.hypot(dx, dy)

    # Ignore very small tracking vibrations
    if movement < NOISE_DEAD_ZONE:
        return old_point

    # Adaptive smoothing
    speed_factor = min(movement / 20.0, 1.0)

    alpha = (
        MIN_SMOOTHING +
        (MAX_SMOOTHING - MIN_SMOOTHING) * speed_factor
    )

    return (
        old_point[0] + alpha * dx,
        old_point[1] + alpha * dy
    )


# ============================================================
# 14. STATIONARY AND MOVEMENT FILTER
# ============================================================

def stabilize_finger(raw_point):

    global stationary_anchor
    global filtered_point
    global movement_active

    clean_point = get_median_point(raw_point)

    # First detected position
    if stationary_anchor is None:

        stationary_anchor = clean_point
        filtered_point = clean_point

        stationary_history.clear()
        stationary_history.append(clean_point)

        return clean_point, False, False

    # --------------------------------------------------------
    # STATIONARY MODE
    # --------------------------------------------------------

    if not movement_active:

        movement = point_distance(
            clean_point,
            stationary_anchor
        )

        if movement < MOVEMENT_START_THRESHOLD:

            return filtered_point, False, False

        # Actual movement started
        movement_active = True

        stationary_history.clear()

        filtered_point = clean_point

        return filtered_point, True, True

    # --------------------------------------------------------
    # ACTIVE MOVEMENT MODE
    # --------------------------------------------------------

    filtered_point = adaptive_smooth(
        clean_point,
        filtered_point
    )

    stationary_history.append(clean_point)

    # --------------------------------------------------------
    # DETECT STATIONARY FINGER
    # --------------------------------------------------------

    if len(stationary_history) >= STATIONARY_FRAMES:

        history = np.array(
            list(stationary_history),
            dtype=np.float32
        )

        center = np.median(
            history,
            axis=0
        )

        distances = np.linalg.norm(
            history - center,
            axis=1
        )

        spread = float(np.max(distances))

        if spread <= STATIONARY_RADIUS:

            movement_active = False

            stationary_anchor = (
                float(center[0]),
                float(center[1])
            )

            filtered_point = stationary_anchor

            stationary_history.clear()

            return filtered_point, False, False

    return filtered_point, True, False


# ============================================================
# 15. CLEAN DRAWING POINTS
# ============================================================

def add_drawing_point(points, new_point):

    if not points:

        points.append(new_point)
        return

    last_point = points[-1]

    distance = point_distance(
        last_point,
        new_point
    )

    if distance >= MIN_POINT_DISTANCE:

        points.append(new_point)


# ============================================================
# 16. CENTRIPETAL CATMULL-ROM CURVE
# ============================================================

def catmull_rom(p0, p1, p2, p3):

    p0 = np.array(p0, dtype=np.float64)
    p1 = np.array(p1, dtype=np.float64)
    p2 = np.array(p2, dtype=np.float64)
    p3 = np.array(p3, dtype=np.float64)

    alpha = 0.5
    epsilon = 1e-4

    def get_t(t, a, b):

        distance = np.linalg.norm(b - a)

        return t + max(distance, epsilon) ** alpha

    t0 = 0.0
    t1 = get_t(t0, p0, p1)
    t2 = get_t(t1, p1, p2)
    t3 = get_t(t2, p2, p3)

    result = []

    for i in range(CURVE_STEPS):

        t = i / CURVE_STEPS

        current_t = t1 + t * (t2 - t1)

        A1 = (
            ((t1 - current_t) / (t1 - t0)) * p0
            +
            ((current_t - t0) / (t1 - t0)) * p1
        )

        A2 = (
            ((t2 - current_t) / (t2 - t1)) * p1
            +
            ((current_t - t1) / (t2 - t1)) * p2
        )

        A3 = (
            ((t3 - current_t) / (t3 - t2)) * p2
            +
            ((current_t - t2) / (t3 - t2)) * p3
        )

        B1 = (
            ((t2 - current_t) / (t2 - t0)) * A1
            +
            ((current_t - t0) / (t2 - t0)) * A2
        )

        B2 = (
            ((t3 - current_t) / (t3 - t1)) * A2
            +
            ((current_t - t1) / (t3 - t1)) * A3
        )

        point = (
            ((t2 - current_t) / (t2 - t1)) * B1
            +
            ((current_t - t1) / (t2 - t1)) * B2
        )

        result.append((
            float(point[0]),
            float(point[1])
        ))

    return result


def get_smooth_curve(points):

    if len(points) <= 2:

        return [
            (
                int(round(p[0])),
                int(round(p[1]))
            )
            for p in points
        ]

    result = []

    extended = (
        [points[0]] +
        points +
        [points[-1]]
    )

    for i in range(1, len(extended) - 2):

        result.extend(
            catmull_rom(
                extended[i - 1],
                extended[i],
                extended[i + 1],
                extended[i + 2]
            )
        )

    result.append((
        int(round(points[-1][0])),
        int(round(points[-1][1]))
    ))

    return [
        (
            int(round(p[0])),
            int(round(p[1]))
        )
        for p in result
    ]


# ============================================================
# 17. DRAWING FUNCTION
# ============================================================

def draw_smooth_stroke(image, points, color, tool="PEN"):

    if not points:
        return

    settings = TOOLS.get(tool, TOOLS["PEN"])

    thickness = settings["thickness"]
    opacity = settings["opacity"]

    radius = max(1, thickness // 2)

    smooth_points = get_smooth_curve(points)

    layer = np.zeros_like(image)

    draw_color = color

    if tool == "PENCIL":

        draw_color = tuple(
            int(c * 0.85 + 255 * 0.15)
            for c in color
        )

    if len(smooth_points) == 1:

        cv2.circle(
            layer,
            smooth_points[0],
            radius,
            draw_color,
            -1,
            cv2.LINE_AA
        )

    else:

        pts = np.array(
            smooth_points,
            dtype=np.int32
        ).reshape((-1, 1, 2))

        cv2.polylines(
            layer,
            [pts],
            False,
            draw_color,
            thickness,
            cv2.LINE_AA
        )

        # Rounded beginning
        cv2.circle(
            layer,
            smooth_points[0],
            radius,
            draw_color,
            -1,
            cv2.LINE_AA
        )

        # Rounded ending
        cv2.circle(
            layer,
            smooth_points[-1],
            radius,
            draw_color,
            -1,
            cv2.LINE_AA
        )

    # Brush softness
    if tool == "BRUSH":

        layer = cv2.GaussianBlur(
            layer,
            (5, 5),
            1.0
        )

    mask = cv2.cvtColor(
        layer,
        cv2.COLOR_BGR2GRAY
    )

    active = mask > 0

    if not np.any(active):
        return

    if opacity >= 1.0:

        image[active] = layer[active]

    else:

        blended = (
            layer.astype(np.float32) * opacity +
            image.astype(np.float32) * (1 - opacity)
        )

        image[active] = np.clip(
            blended[active],
            0,
            255
        ).astype(np.uint8)


# ============================================================
# 18. SAVE STROKE
# ============================================================

def save_current_stroke():

    global current_stroke
    global strokes
    global undo_history

    if not current_stroke:
        return

    stroke = {
        "points": current_stroke.copy(),
        "color": current_stroke_color,
        "tool": current_stroke_tool
    }

    strokes.append(stroke)

    undo_history.append(stroke)
    undo_history = undo_history[-MAX_UNDO:]

    draw_smooth_stroke(
        canvas,
        stroke["points"],
        stroke["color"],
        stroke["tool"]
    )

    current_stroke = []


# ============================================================
# 19. REBUILD CANVAS
# ============================================================

def rebuild_canvas():

    canvas[:] = 255

    for stroke in strokes:

        draw_smooth_stroke(
            canvas,
            stroke["points"],
            stroke["color"],
            stroke["tool"]
        )


# ============================================================
# 20. COLOR PALETTE
# ============================================================

def draw_color_palette(image, selected_index):

    start_x = 20
    start_y = 23

    spacing = 32
    radius = 6

    cv2.rectangle(
        image,
        (8, 7),
        (start_x + 7 * spacing + 12, 48),
        (245, 245, 245),
        -1
    )

    for i, (name, color) in enumerate(COLORS):

        x = start_x + i * spacing
        y = start_y

        cv2.circle(
            image,
            (x, y),
            radius,
            color,
            -1,
            cv2.LINE_AA
        )

        if i == selected_index:

            cv2.circle(
                image,
                (x, y),
                radius + 3,
                (0, 0, 0),
                1,
                cv2.LINE_AA
            )

        cv2.putText(
            image,
            str(i + 1),
            (x - 3, y + 17),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.3,
            (0, 0, 0),
            1,
            cv2.LINE_AA
        )


# ============================================================
# 21. TOOL STATUS
# ============================================================

def draw_tool_status(image):

    text = (
        f"TOOL: {selected_tool} | "
        f"SIZE: {TOOLS[selected_tool]['thickness']}"
    )

    cv2.putText(
        image,
        text,
        (20, 135),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.55,
        (40, 40, 40),
        2,
        cv2.LINE_AA
    )


# ============================================================
# 22. EDGE-AWARE COORDINATE MAPPING
# ============================================================

def map_finger_to_canvas(tip, width, height):

    usable_width = 1.0 - (2.0 * EDGE_MARGIN)
    usable_height = 1.0 - (2.0 * EDGE_MARGIN)

    normalized_x = (
        (tip.x - EDGE_MARGIN) / usable_width
    )

    normalized_y = (
        (tip.y - EDGE_MARGIN) / usable_height
    )

    # Map normalized finger position to canvas
    x = int(round(normalized_x * (width - 1)))
    y = int(round(normalized_y * (height - 1)))

    # Clamp coordinates inside canvas
    x = max(0, min(width - 1, x))
    y = max(0, min(height - 1, y))

    return x, y


# ============================================================
# 23. MAIN APPLICATION
# ============================================================

try:

    with mp_hands.Hands(

        static_image_mode=False,

        max_num_hands=1,

        model_complexity=1,

        min_detection_confidence=MIN_DETECTION_CONFIDENCE,

        min_tracking_confidence=MIN_TRACKING_CONFIDENCE

    ) as hands:

        while True:

            success, frame = cap.read()

            if not success:

                print("Unable to read camera frame.")
                break

            frame = cv2.flip(frame, 1)

            height, width = frame.shape[:2]

            if canvas is None:

                canvas = np.full_like(frame, 255)

            # ----------------------------------------------
            # HAND DETECTION
            # ----------------------------------------------

            small_frame = cv2.resize(
                frame,
                (PROCESS_WIDTH, PROCESS_HEIGHT),
                interpolation=cv2.INTER_AREA
            )

            rgb = cv2.cvtColor(
                small_frame,
                cv2.COLOR_BGR2RGB
            )

            rgb.flags.writeable = False

            results = hands.process(rgb)

            rgb.flags.writeable = True

            hand_detected = bool(
                results.multi_hand_landmarks
            )

            status = "NO HAND DETECTED - SHOW YOUR HAND"
            status_color = (0, 0, 255)

            cursor_point = None

            debug_frame = small_frame.copy()

            # ----------------------------------------------
            # HAND FOUND
            # ----------------------------------------------

            if hand_detected:

                hand = results.multi_hand_landmarks[0]

                mp_drawing.draw_landmarks(
                    debug_frame,
                    hand,
                    mp_hands.HAND_CONNECTIONS
                )

                tip = hand.landmark[8]

                # Improved canvas coordinate mapping
                x, y = map_finger_to_canvas(
                    tip,
                    width,
                    height
                )

                lost_hand_frames = 0

                raw_extended = is_index_finger_extended(hand)

                index_is_extended = update_gesture(
                    raw_extended
                )

                # ------------------------------------------
                # BEFORE START
                # ------------------------------------------

                if not drawing_started:

                    status = "HAND DETECTED - PRESS S TO START"
                    status_color = (0, 150, 0)

                # ------------------------------------------
                # PAUSED
                # ------------------------------------------

                elif drawing_paused:

                    status = "PAUSED - PRESS P TO RESUME"
                    status_color = (0, 0, 255)

                    save_current_stroke()
                    reset_tracking()

                # ------------------------------------------
                # DRAWING MODE
                # ------------------------------------------

                elif index_is_extended:

                    status = "DRAWING"
                    status_color = brush_color

                    stable_point, actual_movement, movement_started = (
                        stabilize_finger((x, y))
                    )

                    smoothed_point = stable_point

                    cursor_point = (
                        int(round(stable_point[0])),
                        int(round(stable_point[1]))
                    )

                    # --------------------------------------
                    # START OR CONTINUE STROKE
                    # --------------------------------------

                    if movement_started:

                        # Only create a new stroke when
                        # the previous stroke does not exist.

                        if not current_stroke:

                            current_stroke_color = brush_color
                            current_stroke_tool = selected_tool

                            start_point = (
                                int(round(stationary_anchor[0])),
                                int(round(stationary_anchor[1]))
                            )

                            current_stroke.append(start_point)

                        # Always synchronize with the last
                        # accepted point after a pause.

                        last_point = current_stroke[-1]

                        previous_point = (
                            float(last_point[0]),
                            float(last_point[1])
                        )

                    # --------------------------------------
                    # ADD DRAWING POINTS
                    # --------------------------------------

                    if (
                        actual_movement
                        and previous_point is not None
                    ):

                        movement = point_distance(
                            stable_point,
                            previous_point
                        )

                        if movement >= MIN_POINT_DISTANCE:

                            add_drawing_point(
                                current_stroke,
                                cursor_point
                            )

                            previous_point = stable_point

                # ------------------------------------------
                # INDEX FINGER NOT EXTENDED
                # ------------------------------------------

                else:

                    status = "HAND DETECTED - EXTEND INDEX FINGER"
                    status_color = (0, 140, 255)

                    save_current_stroke()

                    previous_point = None
                    smoothed_point = None

                    reset_movement_filter()

            # ----------------------------------------------
            # NO HAND DETECTED
            # ----------------------------------------------

            else:

                lost_hand_frames += 1

                if lost_hand_frames >= MAX_LOST_HAND_FRAMES:

                    save_current_stroke()
                    reset_tracking()

            # ----------------------------------------------
            # DEBUG WINDOW
            # ----------------------------------------------

            debug_text = (
                "HAND FOUND" if hand_detected else "NO HAND"
            )

            debug_color = (
                (0, 255, 0) if hand_detected else (0, 0, 255)
            )

            cv2.putText(
                debug_frame,
                debug_text,
                (10, 25),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                debug_color,
                2,
                cv2.LINE_AA
            )

            if SHOW_DEBUG_WINDOW:

                cv2.imshow(
                    "Hand Tracking Debug",
                    debug_frame
                )

            # ----------------------------------------------
            # CANVAS DISPLAY
            # ----------------------------------------------

            output = canvas.copy()

            if current_stroke:

                draw_smooth_stroke(
                    output,
                    current_stroke,
                    current_stroke_color,
                    current_stroke_tool
                )

            draw_color_palette(
                output,
                selected_color_index
            )

            draw_tool_status(output)

            # ----------------------------------------------
            # DRAWING CURSOR
            # ----------------------------------------------

            if (
                cursor_point is not None
                and drawing_started
                and not drawing_paused
            ):

                cv2.circle(
                    output,
                    cursor_point,
                    max(
                        3,
                        TOOLS[selected_tool]["thickness"] // 2
                    ),
                    brush_color,
                    -1,
                    cv2.LINE_AA
                )

            # ----------------------------------------------
            # STATUS
            # ----------------------------------------------

            cv2.putText(
                output,
                status,
                (20, 100),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                status_color,
                2,
                cv2.LINE_AA
            )

            instructions = (
                "S: Start | 1-8: Color | N: Pen | M: Marker | "
                "X: Max | L: Pencil | B: Brush | "
                "P: Pause | Z: Undo | C: Clear | Q: Quit"
            )

            cv2.putText(
                output,
                instructions,
                (20, height - 20),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.43,
                (80, 80, 80),
                1,
                cv2.LINE_AA
            )

            cv2.imshow(
                "Virtual Paint App",
                output
            )

            # ----------------------------------------------
            # KEYBOARD CONTROLS
            # ----------------------------------------------

            key = cv2.waitKey(1) & 0xFF

            if ord("1") <= key <= ord("8"):

                selected_color_index = key - ord("1")

                brush_color = COLORS[selected_color_index][1]

            elif key == ord("n"):

                selected_tool = "PEN"

            elif key == ord("m"):

                selected_tool = "MARKER"

            elif key == ord("x"):

                selected_tool = "MAX THICKNESS"

            elif key == ord("l"):

                selected_tool = "PENCIL"

            elif key == ord("b"):

                selected_tool = "BRUSH"

            elif key == ord("s") and not drawing_started:

                drawing_started = True
                drawing_paused = False

                reset_tracking()

                save_current_stroke()

            elif key == ord("p") and drawing_started:

                drawing_paused = not drawing_paused

                save_current_stroke()
                reset_tracking()

            elif key == ord("z") and drawing_started:

                save_current_stroke()

                if undo_history:

                    last_stroke = undo_history.pop()

                    for i in range(len(strokes) - 1, -1, -1):

                        if strokes[i] is last_stroke:

                            strokes.pop(i)
                            break

                    rebuild_canvas()

                reset_tracking()

            elif key == ord("c"):

                strokes.clear()
                current_stroke.clear()
                undo_history.clear()

                canvas[:] = 255

                drawing_started = False
                drawing_paused = False

                lost_hand_frames = 0

                reset_tracking()

            elif key == ord("q"):

                break

finally:

    cap.release()
    cv2.destroyAllWindows()
```

```python
import cv2
import mediapipe as mp
import numpy as np
import math
from collections import deque

# ============================================================
# 1. CAMERA SETUP
# ============================================================

CAMERA_INDEX = 0
FRAME_WIDTH = 1280
FRAME_HEIGHT = 720

cap = cv2.VideoCapture(CAMERA_INDEX)

if not cap.isOpened():
    cap.release()
    cap = cv2.VideoCapture(1)

if not cap.isOpened():
    raise RuntimeError("Camera not found. Check your camera connection.")

cap.set(cv2.CAP_PROP_FRAME_WIDTH, FRAME_WIDTH)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, FRAME_HEIGHT)
cap.set(cv2.CAP_PROP_FPS, 30)
cap.set(cv2.CAP_PROP_BUFFERSIZE, 1)

mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils

# ============================================================
# 2. COLORS
# ============================================================

COLORS = [
    ("Green",  (0, 200, 0)),
    ("Red",    (0, 0, 255)),
    ("Blue",   (255, 0, 0)),
    ("Black",  (0, 0, 0)),
    ("Yellow", (0, 255, 255)),
    ("Purple", (180, 0, 180)),
    ("Orange", (0, 140, 255)),
    ("Pink",   (203, 192, 255))
]

selected_color_index = 0
brush_color = COLORS[selected_color_index][1]

# ============================================================
# 3. DRAWING TOOLS
# ============================================================

TOOLS = {
    "PEN": {"thickness": 2, "opacity": 1.0},
    "MARKER": {"thickness": 9, "opacity": 0.55},
    "MAX THICKNESS": {"thickness": 24, "opacity": 1.0},
    "PENCIL": {"thickness": 2, "opacity": 0.80},
    "BRUSH": {"thickness": 15, "opacity": 0.90}
}

selected_tool = "PEN"
current_stroke_tool = selected_tool
current_stroke_color = brush_color

# ============================================================
# 4. CANVAS AND APP VARIABLES
# ============================================================

canvas = None

strokes = []
current_stroke = []
undo_history = []

MAX_UNDO = 10

drawing_started = False
drawing_paused = False

previous_point = None
smoothed_point = None

# ============================================================
# 5. HAND DETECTION SETTINGS
# ============================================================

PROCESS_WIDTH = 640
PROCESS_HEIGHT = 360

MIN_DETECTION_CONFIDENCE = 0.50
MIN_TRACKING_CONFIDENCE = 0.50

UP_THRESHOLD = 2
DOWN_THRESHOLD = 3

finger_up_count = 0
finger_down_count = 0
index_is_extended = False

lost_hand_frames = 0
MAX_LOST_HAND_FRAMES = 2

# ============================================================
# 6. ADVANCED DRAWING SETTINGS
# ============================================================

MIN_SMOOTHING = 0.24
MAX_SMOOTHING = 0.62

MIN_POINT_DISTANCE = 1.0

CURVE_STEPS = 14

CURSOR_RADIUS = 6

MOVEMENT_START_THRESHOLD = 3.0
STATIONARY_RADIUS = 1.8
STATIONARY_FRAMES = 7

RAW_HISTORY_SIZE = 3
NOISE_DEAD_ZONE = 0.7

EDGE_MARGIN = 0.035

raw_point_history = deque(maxlen=RAW_HISTORY_SIZE)
stationary_history = deque(maxlen=STATIONARY_FRAMES)

stationary_anchor = None
filtered_point = None
movement_active = False

SHOW_DEBUG_WINDOW = True

ANGLE_THRESHOLD_2D = 125
ANGLE_THRESHOLD_3D = 130
LENGTH_RATIO_THRESHOLD = 1.00

# ============================================================
# 7. DISTANCE FUNCTIONS
# ============================================================

def distance_2d(a, b):

    return math.hypot(
        a.x - b.x,
        a.y - b.y
    )


def distance_3d(a, b):

    return math.sqrt(
        (a.x - b.x) ** 2 +
        (a.y - b.y) ** 2 +
        (a.z - b.z) ** 2
    )


def point_distance(p1, p2):

    return math.hypot(
        p1[0] - p2[0],
        p1[1] - p2[1]
    )


# ============================================================
# 8. ANGLE FUNCTIONS
# ============================================================

def calculate_angle_2d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


def calculate_angle_3d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y, a.z - b.z],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y, c.z - b.z],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


# ============================================================
# 9. INDEX FINGER DETECTION
# ============================================================

def is_index_finger_extended(hand):

    lm = hand.landmark

    mcp = lm[5]
    pip = lm[6]
    dip = lm[7]
    tip = lm[8]

    angle_2d = calculate_angle_2d(mcp, pip, tip)
    angle_3d = calculate_angle_3d(mcp, pip, tip)

    dip_angle_2d = calculate_angle_2d(pip, dip, tip)
    dip_angle_3d = calculate_angle_3d(pip, dip, tip)

    ratio_2d = (
        distance_2d(mcp, tip) /
        max(distance_2d(mcp, pip), 0.0001)
    )

    ratio_3d = (
        distance_3d(mcp, tip) /
        max(distance_3d(mcp, pip), 0.0001)
    )

    angle_ok = (
        angle_2d >= ANGLE_THRESHOLD_2D or
        angle_3d >= ANGLE_THRESHOLD_3D
    )

    length_ok = (
        ratio_2d >= LENGTH_RATIO_THRESHOLD or
        ratio_3d >= LENGTH_RATIO_THRESHOLD
    )

    tip_ok = (
        dip_angle_2d >= 110 or
        dip_angle_3d >= 110
    )

    return angle_ok and length_ok and tip_ok


# ============================================================
# 10. GESTURE STABILIZATION
# ============================================================

def update_gesture(raw_extended):

    global finger_up_count
    global finger_down_count
    global index_is_extended

    if raw_extended:

        finger_up_count += 1
        finger_down_count = 0

    else:

        finger_down_count += 1
        finger_up_count = 0

    if finger_up_count >= UP_THRESHOLD:
        index_is_extended = True

    elif finger_down_count >= DOWN_THRESHOLD:
        index_is_extended = False

    return index_is_extended


# ============================================================
# 11. RESET DRAWING FILTER
# ============================================================

def reset_movement_filter():

    global stationary_anchor
    global filtered_point
    global movement_active

    raw_point_history.clear()
    stationary_history.clear()

    stationary_anchor = None
    filtered_point = None
    movement_active = False


def reset_tracking():

    global previous_point
    global smoothed_point
    global finger_up_count
    global finger_down_count
    global index_is_extended

    previous_point = None
    smoothed_point = None

    finger_up_count = 0
    finger_down_count = 0

    index_is_extended = False

    reset_movement_filter()


# ============================================================
# 12. RAW POINT NOISE REDUCTION
# ============================================================

def get_median_point(raw_point):

    raw_point_history.append(raw_point)

    points = np.array(
        list(raw_point_history),
        dtype=np.float32
    )

    median_x = float(np.median(points[:, 0]))
    median_y = float(np.median(points[:, 1]))

    return median_x, median_y


# ============================================================
# 13. ADAPTIVE SMOOTHING
# ============================================================

def adaptive_smooth(new_point, old_point):

    if old_point is None:

        return (
            float(new_point[0]),
            float(new_point[1])
        )

    dx = new_point[0] - old_point[0]
    dy = new_point[1] - old_point[1]

    movement = math.hypot(dx, dy)

    if movement < NOISE_DEAD_ZONE:
        return old_point

    speed_factor = min(movement / 20.0, 1.0)

    alpha = (
        MIN_SMOOTHING +
        (MAX_SMOOTHING - MIN_SMOOTHING) * speed_factor
    )

    return (
        old_point[0] + alpha * dx,
        old_point[1] + alpha * dy
    )


# ============================================================
# 14. STATIONARY AND MOVEMENT FILTER
# ============================================================

def stabilize_finger(raw_point):

    global stationary_anchor
    global filtered_point
    global movement_active

    clean_point = get_median_point(raw_point)

    if stationary_anchor is None:

        stationary_anchor = clean_point
        filtered_point = clean_point

        stationary_history.clear()
        stationary_history.append(clean_point)

        return clean_point, False, False

    # STATIONARY MODE

    if not movement_active:

        movement = point_distance(
            clean_point,
            stationary_anchor
        )

        if movement < MOVEMENT_START_THRESHOLD:

            return filtered_point, False, False

        movement_active = True

        stationary_history.clear()

        filtered_point = clean_point

        return filtered_point, True, True

    # ACTIVE MOVEMENT MODE

    filtered_point = adaptive_smooth(
        clean_point,
        filtered_point
    )

    stationary_history.append(clean_point)

    # DETECT STATIONARY FINGER

    if len(stationary_history) >= STATIONARY_FRAMES:

        history = np.array(
            list(stationary_history),
            dtype=np.float32
        )

        center = np.median(
            history,
            axis=0
        )

        distances = np.linalg.norm(
            history - center,
            axis=1
        )

        spread = float(np.max(distances))

        if spread <= STATIONARY_RADIUS:

            movement_active = False

            stationary_anchor = (
                float(center[0]),
                float(center[1])
            )

            filtered_point = stationary_anchor

            stationary_history.clear()

            return filtered_point, False, False

    return filtered_point, True, False


# ============================================================
# 15. CLEAN DRAWING POINTS
# ============================================================

def add_drawing_point(points, new_point):

    if not points:

        points.append(new_point)
        return

    last_point = points[-1]

    distance = point_distance(
        last_point,
        new_point
    )

    if distance >= MIN_POINT_DISTANCE:

        points.append(new_point)


# ============================================================
# 16. CENTRIPETAL CATMULL-ROM CURVE
# ============================================================

def catmull_rom(p0, p1, p2, p3):

    p0 = np.array(p0, dtype=np.float64)
    p1 = np.array(p1, dtype=np.float64)
    p2 = np.array(p2, dtype=np.float64)
    p3 = np.array(p3, dtype=np.float64)

    alpha = 0.5
    epsilon = 1e-4

    def get_t(t, a, b):

        distance = np.linalg.norm(b - a)

        return t + max(distance, epsilon) ** alpha

    t0 = 0.0
    t1 = get_t(t0, p0, p1)
    t2 = get_t(t1, p1, p2)
    t3 = get_t(t2, p2, p3)

    result = []

    for i in range(CURVE_STEPS):

        t = i / CURVE_STEPS

        current_t = t1 + t * (t2 - t1)

        A1 = (
            ((t1 - current_t) / (t1 - t0)) * p0
            +
            ((current_t - t0) / (t1 - t0)) * p1
        )

        A2 = (
            ((t2 - current_t) / (t2 - t1)) * p1
            +
            ((current_t - t1) / (t2 - t1)) * p2
        )

        A3 = (
            ((t3 - current_t) / (t3 - t2)) * p2
            +
            ((current_t - t2) / (t3 - t2)) * p3
        )

        B1 = (
            ((t2 - current_t) / (t2 - t0)) * A1
            +
            ((current_t - t0) / (t2 - t0)) * A2
        )

        B2 = (
            ((t3 - current_t) / (t3 - t1)) * A2
            +
            ((current_t - t1) / (t3 - t1)) * A3
        )

        point = (
            ((t2 - current_t) / (t2 - t1)) * B1
            +
            ((current_t - t1) / (t2 - t1)) * B2
        )

        result.append((
            float(point[0]),
            float(point[1])
        ))

    return result


def get_smooth_curve(points):

    if len(points) <= 2:

        return [
            (
                int(round(p[0])),
                int(round(p[1]))
            )
            for p in points
        ]

    result = []

    extended = (
        [points[0]] +
        points +
        [points[-1]]
    )

    for i in range(1, len(extended) - 2):

        result.extend(
            catmull_rom(
                extended[i - 1],
                extended[i],
                extended[i + 1],
                extended[i + 2]
            )
        )

    result.append((
        int(round(points[-1][0])),
        int(round(points[-1][1]))
    ))

    return [
        (
            int(round(p[0])),
            int(round(p[1]))
        )
        for p in result
    ]


# ============================================================
# 17. DRAWING FUNCTION - BLACK COLOR FIXED
# ============================================================

def draw_smooth_stroke(image, points, color, tool="PEN"):

    if not points:
        return

    settings = TOOLS.get(tool, TOOLS["PEN"])

    thickness = settings["thickness"]
    opacity = settings["opacity"]

    radius = max(1, thickness // 2)

    smooth_points = get_smooth_curve(points)

    # Separate color layer and independent drawing mask.
    # This allows pure black (0, 0, 0) to be drawn correctly.
    layer = np.zeros_like(image)
    mask = np.zeros(image.shape[:2], dtype=np.uint8)

    draw_color = color

    if tool == "PENCIL":

        draw_color = tuple(
            int(c * 0.85 + 255 * 0.15)
            for c in color
        )

    # DRAW SINGLE POINT

    if len(smooth_points) == 1:

        cv2.circle(
            layer,
            smooth_points[0],
            radius,
            draw_color,
            -1,
            cv2.LINE_8
        )

        cv2.circle(
            mask,
            smooth_points[0],
            radius,
            255,
            -1,
            cv2.LINE_AA
        )

    # DRAW SMOOTH CURVE

    else:

        pts = np.array(
            smooth_points,
            dtype=np.int32
        ).reshape((-1, 1, 2))

        # Actual colored stroke
        cv2.polylines(
            layer,
            [pts],
            False,
            draw_color,
            thickness,
            cv2.LINE_8
        )

        # Independent antialiased mask
        cv2.polylines(
            mask,
            [pts],
            False,
            255,
            thickness,
            cv2.LINE_AA
        )

        # Rounded beginning
        cv2.circle(
            layer,
            smooth_points[0],
            radius,
            draw_color,
            -1,
            cv2.LINE_8
        )

        cv2.circle(
            mask,
            smooth_points[0],
            radius,
            255,
            -1,
            cv2.LINE_AA
        )

        # Rounded ending
        cv2.circle(
            layer,
            smooth_points[-1],
            radius,
            draw_color,
            -1,
            cv2.LINE_8
        )

        cv2.circle(
            mask,
            smooth_points[-1],
            radius,
            255,
            -1,
            cv2.LINE_AA
        )

    # BRUSH SOFTNESS

    if tool == "BRUSH":

        mask = cv2.GaussianBlur(
            mask,
            (5, 5),
            1.0
        )

    # COLOR AND OPACITY COMPOSITING

    alpha = (
        mask.astype(np.float32) / 255.0
    ) * opacity

    alpha = alpha[:, :, np.newaxis]

    blended = (
        layer.astype(np.float32) * alpha +
        image.astype(np.float32) * (1.0 - alpha)
    )

    image[:] = np.clip(
        blended,
        0,
        255
    ).astype(np.uint8)


# ============================================================
# 18. SAVE STROKE
# ============================================================

def save_current_stroke():

    global current_stroke
    global strokes
    global undo_history

    if not current_stroke:
        return

    stroke = {
        "points": current_stroke.copy(),
        "color": current_stroke_color,
        "tool": current_stroke_tool
    }

    strokes.append(stroke)

    undo_history.append(stroke)
    undo_history = undo_history[-MAX_UNDO:]

    draw_smooth_stroke(
        canvas,
        stroke["points"],
        stroke["color"],
        stroke["tool"]
    )

    current_stroke = []


# ============================================================
# 19. REBUILD CANVAS
# ============================================================

def rebuild_canvas():

    canvas[:] = 255

    for stroke in strokes:

        draw_smooth_stroke(
            canvas,
            stroke["points"],
            stroke["color"],
            stroke["tool"]
        )


# ============================================================
# 20. COLOR PALETTE
# ============================================================

def draw_color_palette(image, selected_index):

    start_x = 20
    start_y = 23

    spacing = 32
    radius = 6

    cv2.rectangle(
        image,
        (8, 7),
        (start_x + 7 * spacing + 12, 48),
        (245, 245, 245),
        -1
    )

    for i, (name, color) in enumerate(COLORS):

        x = start_x + i * spacing
        y = start_y

        cv2.circle(
            image,
            (x, y),
            radius,
            color,
            -1,
            cv2.LINE_AA
        )

        if i == selected_index:

            cv2.circle(
                image,
                (x, y),
                radius + 3,
                (0, 0, 0),
                1,
                cv2.LINE_AA
            )

        cv2.putText(
            image,
            str(i + 1),
            (x - 3, y + 17),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.3,
            (0, 0, 0),
            1,
            cv2.LINE_AA
        )


# ============================================================
# 21. TOOL STATUS
# ============================================================

def draw_tool_status(image):

    text = (
        f"TOOL: {selected_tool} | "
        f"SIZE: {TOOLS[selected_tool]['thickness']}"
    )

    cv2.putText(
        image,
        text,
        (20, 135),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.55,
        (40, 40, 40),
        2,
        cv2.LINE_AA
    )


# ============================================================
# 22. EDGE-AWARE COORDINATE MAPPING
# ============================================================

def map_finger_to_canvas(tip, width, height):

    usable_width = 1.0 - (2.0 * EDGE_MARGIN)
    usable_height = 1.0 - (2.0 * EDGE_MARGIN)

    normalized_x = (
        (tip.x - EDGE_MARGIN) / usable_width
    )

    normalized_y = (
        (tip.y - EDGE_MARGIN) / usable_height
    )

    x = int(round(normalized_x * (width - 1)))
    y = int(round(normalized_y * (height - 1)))

    x = max(0, min(width - 1, x))
    y = max(0, min(height - 1, y))

    return x, y


# ============================================================
# 23. MAIN APPLICATION
# ============================================================

try:

    with mp_hands.Hands(

        static_image_mode=False,

        max_num_hands=1,

        model_complexity=1,

        min_detection_confidence=MIN_DETECTION_CONFIDENCE,

        min_tracking_confidence=MIN_TRACKING_CONFIDENCE

    ) as hands:

        while True:

            success, frame = cap.read()

            if not success:

                print("Unable to read camera frame.")
                break

            frame = cv2.flip(frame, 1)

            height, width = frame.shape[:2]

            if canvas is None:

                canvas = np.full_like(frame, 255)

            # HAND DETECTION

            small_frame = cv2.resize(
                frame,
                (PROCESS_WIDTH, PROCESS_HEIGHT),
                interpolation=cv2.INTER_AREA
            )

            rgb = cv2.cvtColor(
                small_frame,
                cv2.COLOR_BGR2RGB
            )

            rgb.flags.writeable = False

            results = hands.process(rgb)

            rgb.flags.writeable = True

            hand_detected = bool(
                results.multi_hand_landmarks
            )

            status = "NO HAND DETECTED - SHOW YOUR HAND"
            status_color = (0, 0, 255)

            cursor_point = None

            debug_frame = small_frame.copy()

            # HAND FOUND

            if hand_detected:

                hand = results.multi_hand_landmarks[0]

                mp_drawing.draw_landmarks(
                    debug_frame,
                    hand,
                    mp_hands.HAND_CONNECTIONS
                )

                tip = hand.landmark[8]

                x, y = map_finger_to_canvas(
                    tip,
                    width,
                    height
                )

                lost_hand_frames = 0

                raw_extended = is_index_finger_extended(hand)

                index_is_extended = update_gesture(
                    raw_extended
                )

                # BEFORE START

                if not drawing_started:

                    status = "HAND DETECTED - PRESS S TO START"
                    status_color = (0, 150, 0)

                # PAUSED

                elif drawing_paused:

                    status = "PAUSED - PRESS P TO RESUME"
                    status_color = (0, 0, 255)

                    save_current_stroke()
                    reset_tracking()

                # DRAWING MODE

                elif index_is_extended:

                    status = "DRAWING"
                    status_color = brush_color

                    stable_point, actual_movement, movement_started = (
                        stabilize_finger((x, y))
                    )

                    smoothed_point = stable_point

                    cursor_point = (
                        int(round(stable_point[0])),
                        int(round(stable_point[1]))
                    )

                    # START OR CONTINUE STROKE

                    if movement_started:

                        if not current_stroke:

                            current_stroke_color = brush_color
                            current_stroke_tool = selected_tool

                            start_point = (
                                int(round(stationary_anchor[0])),
                                int(round(stationary_anchor[1]))
                            )

                            current_stroke.append(start_point)

                        last_point = current_stroke[-1]

                        previous_point = (
                            float(last_point[0]),
                            float(last_point[1])
                        )

                    # ADD DRAWING POINTS

                    if (
                        actual_movement
                        and previous_point is not None
                    ):

                        movement = point_distance(
                            stable_point,
                            previous_point
                        )

                        if movement >= MIN_POINT_DISTANCE:

                            add_drawing_point(
                                current_stroke,
                                cursor_point
                            )

                            previous_point = stable_point

                # INDEX FINGER NOT EXTENDED

                else:

                    status = "HAND DETECTED - EXTEND INDEX FINGER"
                    status_color = (0, 140, 255)

                    save_current_stroke()

                    previous_point = None
                    smoothed_point = None

                    reset_movement_filter()

            # NO HAND DETECTED

            else:

                lost_hand_frames += 1

                if lost_hand_frames >= MAX_LOST_HAND_FRAMES:

                    save_current_stroke()
                    reset_tracking()

            # DEBUG WINDOW

            debug_text = (
                "HAND FOUND" if hand_detected else "NO HAND"
            )

            debug_color = (
                (0, 255, 0) if hand_detected else (0, 0, 255)
            )

            cv2.putText(
                debug_frame,
                debug_text,
                (10, 25),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                debug_color,
                2,
                cv2.LINE_AA
            )

            if SHOW_DEBUG_WINDOW:

                cv2.imshow(
                    "Hand Tracking Debug",
                    debug_frame
                )

            # CANVAS DISPLAY

            output = canvas.copy()

            if current_stroke:

                draw_smooth_stroke(
                    output,
                    current_stroke,
                    current_stroke_color,
                    current_stroke_tool
                )

            draw_color_palette(
                output,
                selected_color_index
            )

            draw_tool_status(output)

            # DRAWING CURSOR

            if (
                cursor_point is not None
                and drawing_started
                and not drawing_paused
            ):

                cv2.circle(
                    output,
                    cursor_point,
                    max(
                        3,
                        TOOLS[selected_tool]["thickness"] // 2
                    ),
                    brush_color,
                    -1,
                    cv2.LINE_AA
                )

            # STATUS

            cv2.putText(
                output,
                status,
                (20, 100),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                status_color,
                2,
                cv2.LINE_AA
            )

            instructions = (
                "S: Start | 1-8: Color | N: Pen | M: Marker | "
                "X: Max | L: Pencil | B: Brush | "
                "P: Pause | Z: Undo | C: Clear | Q: Quit"
            )

            cv2.putText(
                output,
                instructions,
                (20, height - 20),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.43,
                (80, 80, 80),
                1,
                cv2.LINE_AA
            )

            cv2.imshow(
                "Virtual Paint App",
                output
            )

            # KEYBOARD CONTROLS

            key = cv2.waitKey(1) & 0xFF

            if ord("1") <= key <= ord("8"):

                selected_color_index = key - ord("1")

                brush_color = COLORS[selected_color_index][1]

            elif key == ord("n"):

                selected_tool = "PEN"

            elif key == ord("m"):

                selected_tool = "MARKER"

            elif key == ord("x"):

                selected_tool = "MAX THICKNESS"

            elif key == ord("l"):

                selected_tool = "PENCIL"

            elif key == ord("b"):

                selected_tool = "BRUSH"

            elif key == ord("s") and not drawing_started:

                drawing_started = True
                drawing_paused = False

                reset_tracking()

                save_current_stroke()

            elif key == ord("p") and drawing_started:

                drawing_paused = not drawing_paused

                save_current_stroke()
                reset_tracking()

            elif key == ord("z") and drawing_started:

                save_current_stroke()

                if undo_history:

                    last_stroke = undo_history.pop()

                    for i in range(len(strokes) - 1, -1, -1):

                        if strokes[i] is last_stroke:

                            strokes.pop(i)
                            break

                    rebuild_canvas()

                reset_tracking()

            elif key == ord("c"):

                strokes.clear()
                current_stroke.clear()
                undo_history.clear()

                canvas[:] = 255

                drawing_started = False
                drawing_paused = False

                lost_hand_frames = 0

                reset_tracking()

            elif key == ord("q"):

                break

finally:

    cap.release()
    cv2.destroyAllWindows()
```

```python

import cv2
import mediapipe as mp
import numpy as np
import math
from collections import deque

# ============================================================
# 1. CAMERA SETUP
# ============================================================

CAMERA_INDEX = 0
FRAME_WIDTH = 1280
FRAME_HEIGHT = 720

cap = cv2.VideoCapture(CAMERA_INDEX)

if not cap.isOpened():
    cap.release()
    cap = cv2.VideoCapture(1)

if not cap.isOpened():
    raise RuntimeError("Camera not found. Check your camera connection.")

cap.set(cv2.CAP_PROP_FRAME_WIDTH, FRAME_WIDTH)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, FRAME_HEIGHT)
cap.set(cv2.CAP_PROP_FPS, 30)
cap.set(cv2.CAP_PROP_BUFFERSIZE, 1)

mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils

# ============================================================
# 2. COLORS
# ============================================================

COLORS = [
    ("Green",  (0, 200, 0)),
    ("Red",    (0, 0, 255)),
    ("Blue",   (255, 0, 0)),
    ("Black",  (0, 0, 0)),
    ("Yellow", (0, 255, 255)),
    ("Purple", (180, 0, 180)),
    ("Orange", (0, 140, 255)),
    ("Pink",   (203, 192, 255))
]

selected_color_index = 0
brush_color = COLORS[selected_color_index][1]

# ============================================================
# 3. DRAWING TOOLS
# ============================================================

TOOLS = {
    "PEN": {"thickness": 2, "opacity": 1.0},
    "MARKER": {"thickness": 9, "opacity": 0.55},
    "MAX THICKNESS": {"thickness": 24, "opacity": 1.0},
    "PENCIL": {"thickness": 2, "opacity": 0.80},
    "BRUSH": {"thickness": 15, "opacity": 0.90}
}

selected_tool = "PEN"
current_stroke_tool = selected_tool
current_stroke_color = brush_color

# ============================================================
# 4. CANVAS AND APP VARIABLES
# ============================================================

canvas = None

strokes = []
current_stroke = []
undo_history = []

MAX_UNDO = 10

drawing_started = False
drawing_paused = False

previous_point = None
smoothed_point = None

# ============================================================
# 5. HAND DETECTION SETTINGS
# ============================================================

PROCESS_WIDTH = 640
PROCESS_HEIGHT = 360

MIN_DETECTION_CONFIDENCE = 0.50
MIN_TRACKING_CONFIDENCE = 0.50

UP_THRESHOLD = 2
DOWN_THRESHOLD = 3

finger_up_count = 0
finger_down_count = 0
index_is_extended = False

lost_hand_frames = 0
MAX_LOST_HAND_FRAMES = 2

# ============================================================
# 6. ADVANCED DRAWING SETTINGS
# ============================================================

MIN_SMOOTHING = 0.18
MAX_SMOOTHING = 0.42

MIN_POINT_DISTANCE = 2.0

CURVE_STEPS = 20

CURSOR_RADIUS = 6

MOVEMENT_START_THRESHOLD = 4.0
STATIONARY_RADIUS = 1.5
STATIONARY_FRAMES = 9

RAW_HISTORY_SIZE = 5
NOISE_DEAD_ZONE = 0.45

EDGE_MARGIN = 0.035

raw_point_history = deque(maxlen=RAW_HISTORY_SIZE)
stationary_history = deque(maxlen=STATIONARY_FRAMES)

stationary_anchor = None
filtered_point = None
movement_active = False

SHOW_DEBUG_WINDOW = True

ANGLE_THRESHOLD_2D = 125
ANGLE_THRESHOLD_3D = 130
LENGTH_RATIO_THRESHOLD = 1.00

# ============================================================
# 7. DISTANCE FUNCTIONS
# ============================================================

def distance_2d(a, b):
    return math.hypot(a.x - b.x, a.y - b.y)


def distance_3d(a, b):
    return math.sqrt(
        (a.x - b.x) ** 2 +
        (a.y - b.y) ** 2 +
        (a.z - b.z) ** 2
    )


def point_distance(p1, p2):
    return math.hypot(
        p1[0] - p2[0],
        p1[1] - p2[1]
    )


# ============================================================
# 8. ANGLE FUNCTIONS
# ============================================================

def calculate_angle_2d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


def calculate_angle_3d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y, a.z - b.z],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y, c.z - b.z],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


# ============================================================
# 9. INDEX FINGER DETECTION
# ============================================================

def is_index_finger_extended(hand):

    lm = hand.landmark

    mcp = lm[5]
    pip = lm[6]
    dip = lm[7]
    tip = lm[8]

    angle_2d = calculate_angle_2d(mcp, pip, tip)
    angle_3d = calculate_angle_3d(mcp, pip, tip)

    dip_angle_2d = calculate_angle_2d(pip, dip, tip)
    dip_angle_3d = calculate_angle_3d(pip, dip, tip)

    ratio_2d = (
        distance_2d(mcp, tip) /
        max(distance_2d(mcp, pip), 0.0001)
    )

    ratio_3d = (
        distance_3d(mcp, tip) /
        max(distance_3d(mcp, pip), 0.0001)
    )

    angle_ok = (
        angle_2d >= ANGLE_THRESHOLD_2D or
        angle_3d >= ANGLE_THRESHOLD_3D
    )

    length_ok = (
        ratio_2d >= LENGTH_RATIO_THRESHOLD or
        ratio_3d >= LENGTH_RATIO_THRESHOLD
    )

    tip_ok = (
        dip_angle_2d >= 110 or
        dip_angle_3d >= 110
    )

    return angle_ok and length_ok and tip_ok


# ============================================================
# 10. GESTURE STABILIZATION
# ============================================================

def update_gesture(raw_extended):

    global finger_up_count
    global finger_down_count
    global index_is_extended

    if raw_extended:
        finger_up_count += 1
        finger_down_count = 0
    else:
        finger_down_count += 1
        finger_up_count = 0

    if finger_up_count >= UP_THRESHOLD:
        index_is_extended = True

    elif finger_down_count >= DOWN_THRESHOLD:
        index_is_extended = False

    return index_is_extended


# ============================================================
# 11. RESET DRAWING FILTER
# ============================================================

def reset_movement_filter():

    global stationary_anchor
    global filtered_point
    global movement_active

    raw_point_history.clear()
    stationary_history.clear()

    stationary_anchor = None
    filtered_point = None
    movement_active = False


def reset_tracking():

    global previous_point
    global smoothed_point
    global finger_up_count
    global finger_down_count
    global index_is_extended

    previous_point = None
    smoothed_point = None

    finger_up_count = 0
    finger_down_count = 0

    index_is_extended = False

    reset_movement_filter()


# ============================================================
# 12. RAW POINT NOISE REDUCTION
# ============================================================

def get_median_point(raw_point):

    raw_point_history.append(raw_point)

    points = np.array(
        list(raw_point_history),
        dtype=np.float32
    )

    median_x = float(np.median(points[:, 0]))
    median_y = float(np.median(points[:, 1]))

    return median_x, median_y


# ============================================================
# 13. ADAPTIVE SMOOTHING
# ============================================================

def adaptive_smooth(new_point, old_point):

    if old_point is None:
        return (
            float(new_point[0]),
            float(new_point[1])
        )

    dx = new_point[0] - old_point[0]
    dy = new_point[1] - old_point[1]

    movement = math.hypot(dx, dy)

    if movement < NOISE_DEAD_ZONE:
        return old_point

    speed_factor = min(movement / 30.0, 1.0)

    alpha = (
        MIN_SMOOTHING +
        (MAX_SMOOTHING - MIN_SMOOTHING) * speed_factor
    )

    return (
        old_point[0] + alpha * dx,
        old_point[1] + alpha * dy
    )


# ============================================================
# 14. STATIONARY AND MOVEMENT FILTER
# ============================================================

def stabilize_finger(raw_point):

    global stationary_anchor
    global filtered_point
    global movement_active

    clean_point = get_median_point(raw_point)

    if stationary_anchor is None:

        stationary_anchor = clean_point
        filtered_point = clean_point

        stationary_history.clear()
        stationary_history.append(clean_point)

        return clean_point, False, False

    # STATIONARY MODE

    if not movement_active:

        movement = point_distance(
            clean_point,
            stationary_anchor
        )

        if movement < MOVEMENT_START_THRESHOLD:
            return filtered_point, False, False

        movement_active = True
        stationary_history.clear()

        filtered_point = clean_point

        return filtered_point, True, True

    # ACTIVE MOVEMENT MODE

    filtered_point = adaptive_smooth(
        clean_point,
        filtered_point
    )

    stationary_history.append(clean_point)

    # DETECT STATIONARY FINGER

    if len(stationary_history) >= STATIONARY_FRAMES:

        history = np.array(
            list(stationary_history),
            dtype=np.float32
        )

        center = np.median(history, axis=0)

        distances = np.linalg.norm(
            history - center,
            axis=1
        )

        spread = float(np.max(distances))

        if spread <= STATIONARY_RADIUS:

            movement_active = False

            stationary_anchor = (
                float(center[0]),
                float(center[1])
            )

            filtered_point = stationary_anchor

            stationary_history.clear()

            return filtered_point, False, False

    return filtered_point, True, False


# ============================================================
# 15. CLEAN DRAWING POINTS
# ============================================================

def add_drawing_point(points, new_point):

    if not points:
        points.append((
            float(new_point[0]),
            float(new_point[1])
        ))
        return

    last_point = points[-1]

    distance = point_distance(last_point, new_point)

    if distance >= MIN_POINT_DISTANCE:
        points.append((
            float(new_point[0]),
            float(new_point[1])
        ))


# ============================================================
# 16. CENTRIPETAL CATMULL-ROM CURVE
# ============================================================

def catmull_rom(p0, p1, p2, p3):

    p0 = np.array(p0, dtype=np.float64)
    p1 = np.array(p1, dtype=np.float64)
    p2 = np.array(p2, dtype=np.float64)
    p3 = np.array(p3, dtype=np.float64)

    alpha = 0.5
    epsilon = 1e-4

    def get_t(t, a, b):

        distance = np.linalg.norm(b - a)

        return t + max(distance, epsilon) ** alpha

    t0 = 0.0
    t1 = get_t(t0, p0, p1)
    t2 = get_t(t1, p1, p2)
    t3 = get_t(t2, p2, p3)

    result = []

    for i in range(CURVE_STEPS + 1):

        t = i / CURVE_STEPS
        current_t = t1 + t * (t2 - t1)

        A1 = (
            ((t1 - current_t) / (t1 - t0)) * p0
            +
            ((current_t - t0) / (t1 - t0)) * p1
        )

        A2 = (
            ((t2 - current_t) / (t2 - t1)) * p1
            +
            ((current_t - t1) / (t2 - t1)) * p2
        )

        A3 = (
            ((t3 - current_t) / (t3 - t2)) * p2
            +
            ((current_t - t2) / (t3 - t2)) * p3
        )

        B1 = (
            ((t2 - current_t) / (t2 - t0)) * A1
            +
            ((current_t - t0) / (t2 - t0)) * A2
        )

        B2 = (
            ((t3 - current_t) / (t3 - t1)) * A2
            +
            ((current_t - t1) / (t3 - t1)) * A3
        )

        point = (
            ((t2 - current_t) / (t2 - t1)) * B1
            +
            ((current_t - t1) / (t2 - t1)) * B2
        )

        result.append((
            float(point[0]),
            float(point[1])
        ))

    return result


def get_smooth_curve(points):

    if len(points) <= 2:
        return [
            (
                int(round(p[0])),
                int(round(p[1]))
            )
            for p in points
        ]

    result = []

    extended = (
        [points[0]] +
        points +
        [points[-1]]
    )

    for i in range(1, len(extended) - 2):

        curve_points = catmull_rom(
            extended[i - 1],
            extended[i],
            extended[i + 1],
            extended[i + 2]
        )

        if result:
            curve_points = curve_points[1:]

        result.extend(curve_points)

    return [
        (
            int(round(p[0])),
            int(round(p[1]))
        )
        for p in result
    ]


# ============================================================
# 17. DRAWING FUNCTION - BLACK COLOR FIXED
# ============================================================

def draw_smooth_stroke(image, points, color, tool="PEN"):

    if not points:
        return

    settings = TOOLS.get(tool, TOOLS["PEN"])

    thickness = settings["thickness"]
    opacity = settings["opacity"]

    radius = max(1, thickness // 2)

    smooth_points = get_smooth_curve(points)

    layer = np.zeros_like(image)
    mask = np.zeros(image.shape[:2], dtype=np.uint8)

    draw_color = color

    if tool == "PENCIL":
        draw_color = tuple(
            int(c * 0.85 + 255 * 0.15)
            for c in color
        )

    if len(smooth_points) == 1:

        cv2.circle(
            layer,
            smooth_points[0],
            radius,
            draw_color,
            -1,
            cv2.LINE_AA
        )

        cv2.circle(
            mask,
            smooth_points[0],
            radius,
            255,
            -1,
            cv2.LINE_AA
        )

    else:

        pts = np.array(
            smooth_points,
            dtype=np.int32
        ).reshape((-1, 1, 2))

        cv2.polylines(
            layer,
            [pts],
            False,
            draw_color,
            thickness,
            cv2.LINE_AA
        )

        cv2.polylines(
            mask,
            [pts],
            False,
            255,
            thickness,
            cv2.LINE_AA
        )

        # Rounded ends

        for point in (
            smooth_points[0],
            smooth_points[-1]
        ):

            cv2.circle(
                layer,
                point,
                radius,
                draw_color,
                -1,
                cv2.LINE_AA
            )

            cv2.circle(
                mask,
                point,
                radius,
                255,
                -1,
                cv2.LINE_AA
            )

    # BRUSH SOFTNESS

    if tool == "BRUSH":

        mask = cv2.GaussianBlur(
            mask,
            (5, 5),
            1.0
        )

    # COLOR AND OPACITY COMPOSITING

    alpha = (
        mask.astype(np.float32) / 255.0
    ) * opacity

    alpha = alpha[:, :, np.newaxis]

    blended = (
        layer.astype(np.float32) * alpha +
        image.astype(np.float32) * (1.0 - alpha)
    )

    image[:] = np.clip(
        blended,
        0,
        255
    ).astype(np.uint8)


# ============================================================
# 18. SAVE STROKE
# ============================================================

def save_current_stroke():

    global current_stroke
    global strokes
    global undo_history

    if not current_stroke:
        return

    stroke = {
        "points": current_stroke.copy(),
        "color": current_stroke_color,
        "tool": current_stroke_tool
    }

    strokes.append(stroke)

    undo_history.append(stroke)
    undo_history = undo_history[-MAX_UNDO:]

    draw_smooth_stroke(
        canvas,
        stroke["points"],
        stroke["color"],
        stroke["tool"]
    )

    current_stroke = []


# ============================================================
# 19. REBUILD CANVAS
# ============================================================

def rebuild_canvas():

    canvas[:] = 255

    for stroke in strokes:

        draw_smooth_stroke(
            canvas,
            stroke["points"],
            stroke["color"],
            stroke["tool"]
        )


# ============================================================
# 20. COLOR PALETTE
# ============================================================

def draw_color_palette(image, selected_index):

    start_x = 20
    start_y = 23

    spacing = 32
    radius = 6

    cv2.rectangle(
        image,
        (8, 7),
        (start_x + 7 * spacing + 12, 48),
        (245, 245, 245),
        -1
    )

    for i, (name, color) in enumerate(COLORS):

        x = start_x + i * spacing
        y = start_y

        cv2.circle(
            image,
            (x, y),
            radius,
            color,
            -1,
            cv2.LINE_AA
        )

        if i == selected_index:

            cv2.circle(
                image,
                (x, y),
                radius + 3,
                (0, 0, 0),
                1,
                cv2.LINE_AA
            )

        cv2.putText(
            image,
            str(i + 1),
            (x - 3, y + 17),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.3,
            (0, 0, 0),
            1,
            cv2.LINE_AA
        )


# ============================================================
# 21. TOOL STATUS
# ============================================================

def draw_tool_status(image):

    text = (
        f"TOOL: {selected_tool} | "
        f"SIZE: {TOOLS[selected_tool]['thickness']}"
    )

    cv2.putText(
        image,
        text,
        (20, 135),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.55,
        (40, 40, 40),
        2,
        cv2.LINE_AA
    )


# ============================================================
# 22. EDGE-AWARE COORDINATE MAPPING
# ============================================================

def map_finger_to_canvas(tip, width, height):

    usable_width = 1.0 - (2.0 * EDGE_MARGIN)
    usable_height = 1.0 - (2.0 * EDGE_MARGIN)

    normalized_x = (
        (tip.x - EDGE_MARGIN) / usable_width
    )

    normalized_y = (
        (tip.y - EDGE_MARGIN) / usable_height
    )

    x = int(round(normalized_x * (width - 1)))
    y = int(round(normalized_y * (height - 1)))

    x = max(0, min(width - 1, x))
    y = max(0, min(height - 1, y))

    return x, y


# ============================================================
# 23. MAIN APPLICATION
# ============================================================

try:

    with mp_hands.Hands(

        static_image_mode=False,
        max_num_hands=1,
        model_complexity=1,
        min_detection_confidence=MIN_DETECTION_CONFIDENCE,
        min_tracking_confidence=MIN_TRACKING_CONFIDENCE

    ) as hands:

        while True:

            success, frame = cap.read()

            if not success:
                print("Unable to read camera frame.")
                break

            frame = cv2.flip(frame, 1)

            height, width = frame.shape[:2]

            if canvas is None:
                canvas = np.full_like(frame, 255)

            # HAND DETECTION

            small_frame = cv2.resize(
                frame,
                (PROCESS_WIDTH, PROCESS_HEIGHT),
                interpolation=cv2.INTER_AREA
            )

            rgb = cv2.cvtColor(
                small_frame,
                cv2.COLOR_BGR2RGB
            )

            rgb.flags.writeable = False
            results = hands.process(rgb)
            rgb.flags.writeable = True

            hand_detected = bool(
                results.multi_hand_landmarks
            )

            status = "NO HAND DETECTED - SHOW YOUR HAND"
            status_color = (0, 0, 255)

            cursor_point = None

            debug_frame = small_frame.copy()

            # HAND FOUND

            if hand_detected:

                hand = results.multi_hand_landmarks[0]

                mp_drawing.draw_landmarks(
                    debug_frame,
                    hand,
                    mp_hands.HAND_CONNECTIONS
                )

                tip = hand.landmark[8]

                x, y = map_finger_to_canvas(
                    tip,
                    width,
                    height
                )

                lost_hand_frames = 0

                raw_extended = is_index_finger_extended(hand)

                index_is_extended = update_gesture(
                    raw_extended
                )

                # BEFORE START

                if not drawing_started:

                    status = "HAND DETECTED - PRESS S TO START"
                    status_color = (0, 150, 0)

                # PAUSED

                elif drawing_paused:

                    status = "PAUSED - PRESS P TO RESUME"
                    status_color = (0, 0, 255)

                    save_current_stroke()
                    reset_tracking()

                # DRAWING MODE

                elif index_is_extended:

                    status = "DRAWING"
                    status_color = brush_color

                    stable_point, actual_movement, movement_started = (
                        stabilize_finger((x, y))
                    )

                    smoothed_point = stable_point

                    cursor_point = (
                        int(round(stable_point[0])),
                        int(round(stable_point[1]))
                    )

                    # START OR CONTINUE STROKE

                    if movement_started:

                        if not current_stroke:

                            current_stroke_color = brush_color
                            current_stroke_tool = selected_tool

                            start_point = (
                                float(stationary_anchor[0]),
                                float(stationary_anchor[1])
                            )

                            current_stroke.append(start_point)

                        last_point = current_stroke[-1]

                        previous_point = (
                            float(last_point[0]),
                            float(last_point[1])
                        )

                    # ADD DRAWING POINTS

                    if (
                        actual_movement
                        and previous_point is not None
                    ):

                        movement = point_distance(
                            stable_point,
                            previous_point
                        )

                        if movement >= MIN_POINT_DISTANCE:

                            add_drawing_point(
                                current_stroke,
                                stable_point
                            )

                            previous_point = stable_point

                # INDEX FINGER NOT EXTENDED

                else:

                    status = "HAND DETECTED - EXTEND INDEX FINGER"
                    status_color = (0, 140, 255)

                    save_current_stroke()

                    previous_point = None
                    smoothed_point = None

                    reset_movement_filter()

            # NO HAND DETECTED

            else:

                lost_hand_frames += 1

                if lost_hand_frames >= MAX_LOST_HAND_FRAMES:

                    save_current_stroke()
                    reset_tracking()

            # DEBUG WINDOW

            debug_text = (
                "HAND FOUND" if hand_detected else "NO HAND"
            )

            debug_color = (
                (0, 255, 0) if hand_detected else (0, 0, 255)
            )

            cv2.putText(
                debug_frame,
                debug_text,
                (10, 25),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                debug_color,
                2,
                cv2.LINE_AA
            )

            if SHOW_DEBUG_WINDOW:

                cv2.imshow(
                    "Hand Tracking Debug",
                    debug_frame
                )

            # CANVAS DISPLAY

            output = canvas.copy()

            if current_stroke:

                draw_smooth_stroke(
                    output,
                    current_stroke,
                    current_stroke_color,
                    current_stroke_tool
                )

            draw_color_palette(
                output,
                selected_color_index
            )

            draw_tool_status(output)

            # DRAWING CURSOR

            if (
                cursor_point is not None
                and drawing_started
                and not drawing_paused
            ):

                cv2.circle(
                    output,
                    cursor_point,
                    max(
                        3,
                        TOOLS[selected_tool]["thickness"] // 2
                    ),
                    brush_color,
                    -1,
                    cv2.LINE_AA
                )

            # STATUS

            cv2.putText(
                output,
                status,
                (20, 100),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                status_color,
                2,
                cv2.LINE_AA
            )

            instructions = (
                "S: Start | 1-8: Color | N: Pen | M: Marker | "
                "X: Max | L: Pencil | B: Brush | "
                "P: Pause | Z: Undo | C: Clear | Q: Quit"
            )

            cv2.putText(
                output,
                instructions,
                (20, height - 20),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.43,
                (80, 80, 80),
                1,
                cv2.LINE_AA
            )

            cv2.imshow(
                "Virtual Paint App",
                output
            )

            # KEYBOARD CONTROLS

            key = cv2.waitKey(1) & 0xFF

            if ord("1") <= key <= ord("8"):

                selected_color_index = key - ord("1")
                brush_color = COLORS[selected_color_index][1]

            elif key == ord("n"):

                selected_tool = "PEN"

            elif key == ord("m"):

                selected_tool = "MARKER"

            elif key == ord("x"):

                selected_tool = "MAX THICKNESS"

            elif key == ord("l"):

                selected_tool = "PENCIL"

            elif key == ord("b"):

                selected_tool = "BRUSH"

            elif key == ord("s") and not drawing_started:

                drawing_started = True
                drawing_paused = False

                reset_tracking()
                save_current_stroke()

            elif key == ord("p") and drawing_started:

                drawing_paused = not drawing_paused

                save_current_stroke()
                reset_tracking()

            elif key == ord("z") and drawing_started:

                save_current_stroke()

                if undo_history:

                    last_stroke = undo_history.pop()

                    for i in range(len(strokes) - 1, -1, -1):

                        if strokes[i] is last_stroke:

                            strokes.pop(i)
                            break

                    rebuild_canvas()

                reset_tracking()

            elif key == ord("c"):

                strokes.clear()
                current_stroke.clear()
                undo_history.clear()

                canvas[:] = 255

                drawing_started = False
                drawing_paused = False

                lost_hand_frames = 0

                reset_tracking()

            elif key == ord("q"):

                break

finally:

    cap.release()
    cv2.destroyAllWindows()
```

```python

import cv2
import mediapipe as mp
import numpy as np
import math
from collections import deque

# ============================================================
# 1. CAMERA SETUP
# ============================================================

CAMERA_INDEX = 0
FRAME_WIDTH = 1280
FRAME_HEIGHT = 720

cap = cv2.VideoCapture(CAMERA_INDEX)

if not cap.isOpened():
    cap.release()
    cap = cv2.VideoCapture(1)

if not cap.isOpened():
    raise RuntimeError("Camera not found. Check your camera connection.")

cap.set(cv2.CAP_PROP_FRAME_WIDTH, FRAME_WIDTH)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, FRAME_HEIGHT)
cap.set(cv2.CAP_PROP_FPS, 30)
cap.set(cv2.CAP_PROP_BUFFERSIZE, 1)

mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils

# ============================================================
# 2. COLORS
# ============================================================

COLORS = [
    ("Green",  (0, 200, 0)),
    ("Red",    (0, 0, 255)),
    ("Blue",   (255, 0, 0)),
    ("Black",  (0, 0, 0)),
    ("Yellow", (0, 255, 255)),
    ("Purple", (180, 0, 180)),
    ("Orange", (0, 140, 255)),
    ("Pink",   (203, 192, 255))
]

selected_color_index = 0
brush_color = COLORS[selected_color_index][1]

# ============================================================
# 3. DRAWING TOOLS
# ============================================================

TOOLS = {
    "PEN": {"thickness": 2, "opacity": 1.0},
    "MARKER": {"thickness": 9, "opacity": 0.55},
    "MAX THICKNESS": {"thickness": 24, "opacity": 1.0},
    "PENCIL": {"thickness": 2, "opacity": 0.80},
    "BRUSH": {"thickness": 15, "opacity": 0.90}
}

selected_tool = "PEN"
current_stroke_tool = selected_tool
current_stroke_color = brush_color

# ============================================================
# 4. CANVAS AND APP VARIABLES
# ============================================================

canvas = None

strokes = []
current_stroke = []
undo_history = []

MAX_UNDO = 10

drawing_started = False
drawing_paused = False

previous_point = None
smoothed_point = None

# ============================================================
# 5. HAND DETECTION SETTINGS
# ============================================================

PROCESS_WIDTH = 640
PROCESS_HEIGHT = 360

MIN_DETECTION_CONFIDENCE = 0.50
MIN_TRACKING_CONFIDENCE = 0.50

UP_THRESHOLD = 2
DOWN_THRESHOLD = 3

finger_up_count = 0
finger_down_count = 0
index_is_extended = False

lost_hand_frames = 0
MAX_LOST_HAND_FRAMES = 2

# ============================================================
# 6. DRAWING QUALITY SETTINGS
# ============================================================

# Smoothing
MIN_SMOOTHING = 0.32
MAX_SMOOTHING = 0.62

# Keep small movements for better circles
MIN_POINT_DISTANCE = 1.0

# Curve quality
CURVE_STEPS = 12

# Cursor
CURSOR_RADIUS = 6

# Movement detection
MOVEMENT_START_THRESHOLD = 3.0
STATIONARY_RADIUS = 1.2
STATIONARY_FRAMES = 12

# Noise filtering
RAW_HISTORY_SIZE = 3
NOISE_DEAD_ZONE = 0.30

EDGE_MARGIN = 0.035

raw_point_history = deque(maxlen=RAW_HISTORY_SIZE)
stationary_history = deque(maxlen=STATIONARY_FRAMES)

stationary_anchor = None
filtered_point = None
movement_active = False

SHOW_DEBUG_WINDOW = True

ANGLE_THRESHOLD_2D = 125
ANGLE_THRESHOLD_3D = 130
LENGTH_RATIO_THRESHOLD = 1.00

# ============================================================
# 7. DISTANCE FUNCTIONS
# ============================================================

def distance_2d(a, b):
    return math.hypot(a.x - b.x, a.y - b.y)


def distance_3d(a, b):
    return math.sqrt(
        (a.x - b.x) ** 2 +
        (a.y - b.y) ** 2 +
        (a.z - b.z) ** 2
    )


def point_distance(p1, p2):
    return math.hypot(
        p1[0] - p2[0],
        p1[1] - p2[1]
    )


# ============================================================
# 8. ANGLE FUNCTIONS
# ============================================================

def calculate_angle_2d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


def calculate_angle_3d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y, a.z - b.z],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y, c.z - b.z],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


# ============================================================
# 9. INDEX FINGER DETECTION
# ============================================================

def is_index_finger_extended(hand):

    lm = hand.landmark

    mcp = lm[5]
    pip = lm[6]
    dip = lm[7]
    tip = lm[8]

    angle_2d = calculate_angle_2d(mcp, pip, tip)
    angle_3d = calculate_angle_3d(mcp, pip, tip)

    dip_angle_2d = calculate_angle_2d(pip, dip, tip)
    dip_angle_3d = calculate_angle_3d(pip, dip, tip)

    ratio_2d = (
        distance_2d(mcp, tip) /
        max(distance_2d(mcp, pip), 0.0001)
    )

    ratio_3d = (
        distance_3d(mcp, tip) /
        max(distance_3d(mcp, pip), 0.0001)
    )

    angle_ok = (
        angle_2d >= ANGLE_THRESHOLD_2D or
        angle_3d >= ANGLE_THRESHOLD_3D
    )

    length_ok = (
        ratio_2d >= LENGTH_RATIO_THRESHOLD or
        ratio_3d >= LENGTH_RATIO_THRESHOLD
    )

    tip_ok = (
        dip_angle_2d >= 110 or
        dip_angle_3d >= 110
    )

    return angle_ok and length_ok and tip_ok


# ============================================================
# 10. GESTURE STABILIZATION
# ============================================================

def update_gesture(raw_extended):

    global finger_up_count
    global finger_down_count
    global index_is_extended

    if raw_extended:
        finger_up_count += 1
        finger_down_count = 0
    else:
        finger_down_count += 1
        finger_up_count = 0

    if finger_up_count >= UP_THRESHOLD:
        index_is_extended = True

    elif finger_down_count >= DOWN_THRESHOLD:
        index_is_extended = False

    return index_is_extended


# ============================================================
# 11. RESET DRAWING FILTER
# ============================================================

def reset_movement_filter():

    global stationary_anchor
    global filtered_point
    global movement_active

    raw_point_history.clear()
    stationary_history.clear()

    stationary_anchor = None
    filtered_point = None
    movement_active = False


def reset_tracking():

    global previous_point
    global smoothed_point
    global finger_up_count
    global finger_down_count
    global index_is_extended

    previous_point = None
    smoothed_point = None

    finger_up_count = 0
    finger_down_count = 0

    index_is_extended = False

    reset_movement_filter()


# ============================================================
# 12. RAW POINT NOISE REDUCTION
# ============================================================

def get_median_point(raw_point):

    raw_point_history.append(raw_point)

    points = np.array(
        list(raw_point_history),
        dtype=np.float32
    )

    return (
        float(np.median(points[:, 0])),
        float(np.median(points[:, 1]))
    )


# ============================================================
# 13. ADAPTIVE SMOOTHING
# ============================================================

def adaptive_smooth(new_point, old_point):

    if old_point is None:
        return (
            float(new_point[0]),
            float(new_point[1])
        )

    dx = new_point[0] - old_point[0]
    dy = new_point[1] - old_point[1]

    movement = math.hypot(dx, dy)

    if movement < NOISE_DEAD_ZONE:
        return old_point

    # Faster movement gets a higher alpha.
    # This reduces lag while preserving smooth movement.

    speed_factor = min(movement / 25.0, 1.0)

    alpha = (
        MIN_SMOOTHING +
        (MAX_SMOOTHING - MIN_SMOOTHING) * speed_factor
    )

    return (
        old_point[0] + alpha * dx,
        old_point[1] + alpha * dy
    )


# ============================================================
# 14. STATIONARY AND MOVEMENT FILTER
# ============================================================

def stabilize_finger(raw_point):

    global stationary_anchor
    global filtered_point
    global movement_active

    clean_point = get_median_point(raw_point)

    if stationary_anchor is None:

        stationary_anchor = clean_point
        filtered_point = clean_point

        stationary_history.clear()
        stationary_history.append(clean_point)

        return clean_point, False, False

    # STATIONARY MODE

    if not movement_active:

        movement = point_distance(
            clean_point,
            stationary_anchor
        )

        if movement < MOVEMENT_START_THRESHOLD:
            return filtered_point, False, False

        movement_active = True

        stationary_history.clear()

        filtered_point = clean_point

        return filtered_point, True, True

    # ACTIVE MOVEMENT MODE

    filtered_point = adaptive_smooth(
        clean_point,
        filtered_point
    )

    stationary_history.append(clean_point)

    # DETECT STATIONARY FINGER

    if len(stationary_history) >= STATIONARY_FRAMES:

        history = np.array(
            list(stationary_history),
            dtype=np.float32
        )

        center = np.median(history, axis=0)

        distances = np.linalg.norm(
            history - center,
            axis=1
        )

        spread = float(np.max(distances))

        if spread <= STATIONARY_RADIUS:

            movement_active = False

            stationary_anchor = (
                float(center[0]),
                float(center[1])
            )

            filtered_point = stationary_anchor

            stationary_history.clear()

            return filtered_point, False, False

    return filtered_point, True, False


# ============================================================
# 15. CLEAN DRAWING POINTS
# ============================================================

def add_drawing_point(points, new_point):

    if not points:
        points.append((
            float(new_point[0]),
            float(new_point[1])
        ))
        return

    last_point = points[-1]

    distance = point_distance(last_point, new_point)

    if distance >= MIN_POINT_DISTANCE:

        points.append((
            float(new_point[0]),
            float(new_point[1])
        ))


# ============================================================
# 16. SAFE CENTRIPETAL CATMULL-ROM CURVE
# ============================================================

def catmull_rom(p0, p1, p2, p3):

    p0 = np.array(p0, dtype=np.float64)
    p1 = np.array(p1, dtype=np.float64)
    p2 = np.array(p2, dtype=np.float64)
    p3 = np.array(p3, dtype=np.float64)

    alpha = 0.5
    epsilon = 1e-4

    def get_t(t, a, b):

        distance = np.linalg.norm(b - a)

        return t + max(distance, epsilon) ** alpha

    t0 = 0.0
    t1 = get_t(t0, p0, p1)
    t2 = get_t(t1, p1, p2)
    t3 = get_t(t2, p2, p3)

    result = []

    for i in range(CURVE_STEPS):

        t = i / CURVE_STEPS
        current_t = t1 + t * (t2 - t1)

        A1 = (
            ((t1 - current_t) / (t1 - t0)) * p0
            +
            ((current_t - t0) / (t1 - t0)) * p1
        )

        A2 = (
            ((t2 - current_t) / (t2 - t1)) * p1
            +
            ((current_t - t1) / (t2 - t1)) * p2
        )

        A3 = (
            ((t3 - current_t) / (t3 - t2)) * p2
            +
            ((current_t - t2) / (t3 - t2)) * p3
        )

        B1 = (
            ((t2 - current_t) / (t2 - t0)) * A1
            +
            ((current_t - t0) / (t2 - t0)) * A2
        )

        B2 = (
            ((t3 - current_t) / (t3 - t1)) * A2
            +
            ((current_t - t1) / (t3 - t1)) * A3
        )

        point = (
            ((t2 - current_t) / (t2 - t1)) * B1
            +
            ((current_t - t1) / (t2 - t1)) * B2
        )

        # Prevent excessive curve overshooting.
        # A small margin preserves natural rounded curves.

        lower = np.minimum(p1, p2) - 2.0
        upper = np.maximum(p1, p2) + 2.0

        point = np.clip(point, lower, upper)

        result.append((
            float(point[0]),
            float(point[1])
        ))

    return result


def get_smooth_curve(points):

    if len(points) <= 2:

        return [
            (
                int(round(p[0])),
                int(round(p[1]))
            )
            for p in points
        ]

    result = []

    extended = (
        [points[0]] +
        points +
        [points[-1]]
    )

    for i in range(1, len(extended) - 2):

        curve_points = catmull_rom(
            extended[i - 1],
            extended[i],
            extended[i + 1],
            extended[i + 2]
        )

        if result:
            curve_points = curve_points[1:]

        result.extend(curve_points)

    result.append(points[-1])

    return [
        (
            int(round(p[0])),
            int(round(p[1]))
        )
        for p in result
    ]


# ============================================================
# 17. HIGH QUALITY DRAWING RENDERING
# ============================================================

def draw_smooth_stroke(image, points, color, tool="PEN"):

    if not points:
        return

    settings = TOOLS.get(tool, TOOLS["PEN"])

    thickness = settings["thickness"]
    opacity = settings["opacity"]

    radius = max(1, thickness // 2)

    smooth_points = get_smooth_curve(points)

    # Independent color layer and mask.
    # This keeps black color working correctly.

    layer = np.zeros_like(image)
    mask = np.zeros(image.shape[:2], dtype=np.uint8)

    draw_color = color

    if tool == "PENCIL":

        draw_color = tuple(
            int(c * 0.85 + 255 * 0.15)
            for c in color
        )

    # SINGLE POINT

    if len(smooth_points) == 1:

        cv2.circle(
            layer,
            smooth_points[0],
            radius,
            draw_color,
            -1,
            cv2.LINE_AA
        )

        cv2.circle(
            mask,
            smooth_points[0],
            radius,
            255,
            -1,
            cv2.LINE_AA
        )

    # SMOOTH CURVE

    else:

        pts = np.array(
            smooth_points,
            dtype=np.int32
        ).reshape((-1, 1, 2))

        cv2.polylines(
            layer,
            [pts],
            False,
            draw_color,
            thickness,
            cv2.LINE_AA
        )

        cv2.polylines(
            mask,
            [pts],
            False,
            255,
            thickness,
            cv2.LINE_AA
        )

        # Rounded line caps

        for point in (
            smooth_points[0],
            smooth_points[-1]
        ):

            cv2.circle(
                layer,
                point,
                radius,
                draw_color,
                -1,
                cv2.LINE_AA
            )

            cv2.circle(
                mask,
                point,
                radius,
                255,
                -1,
                cv2.LINE_AA
            )

    # BRUSH SOFTNESS

    if tool == "BRUSH":

        mask = cv2.GaussianBlur(
            mask,
            (5, 5),
            1.0
        )

    # ALPHA COMPOSITING

    alpha = (
        mask.astype(np.float32) / 255.0
    ) * opacity

    alpha = alpha[:, :, np.newaxis]

    blended = (
        layer.astype(np.float32) * alpha +
        image.astype(np.float32) * (1.0 - alpha)
    )

    image[:] = np.clip(
        blended,
        0,
        255
    ).astype(np.uint8)


# ============================================================
# 18. SAVE STROKE
# ============================================================

def save_current_stroke():

    global current_stroke
    global strokes
    global undo_history

    if not current_stroke:
        return

    stroke = {
        "points": current_stroke.copy(),
        "color": current_stroke_color,
        "tool": current_stroke_tool
    }

    strokes.append(stroke)

    undo_history.append(stroke)
    undo_history = undo_history[-MAX_UNDO:]

    draw_smooth_stroke(
        canvas,
        stroke["points"],
        stroke["color"],
        stroke["tool"]
    )

    current_stroke = []


# ============================================================
# 19. REBUILD CANVAS
# ============================================================

def rebuild_canvas():

    canvas[:] = 255

    for stroke in strokes:

        draw_smooth_stroke(
            canvas,
            stroke["points"],
            stroke["color"],
            stroke["tool"]
        )


# ============================================================
# 20. COLOR PALETTE
# ============================================================

def draw_color_palette(image, selected_index):

    start_x = 20
    start_y = 23

    spacing = 32
    radius = 6

    cv2.rectangle(
        image,
        (8, 7),
        (start_x + 7 * spacing + 12, 48),
        (245, 245, 245),
        -1
    )

    for i, (name, color) in enumerate(COLORS):

        x = start_x + i * spacing
        y = start_y

        cv2.circle(
            image,
            (x, y),
            radius,
            color,
            -1,
            cv2.LINE_AA
        )

        if i == selected_index:

            cv2.circle(
                image,
                (x, y),
                radius + 3,
                (0, 0, 0),
                1,
                cv2.LINE_AA
            )

        cv2.putText(
            image,
            str(i + 1),
            (x - 3, y + 17),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.3,
            (0, 0, 0),
            1,
            cv2.LINE_AA
        )


# ============================================================
# 21. TOOL STATUS
# ============================================================

def draw_tool_status(image):

    text = (
        f"TOOL: {selected_tool} | "
        f"SIZE: {TOOLS[selected_tool]['thickness']}"
    )

    cv2.putText(
        image,
        text,
        (20, 135),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.55,
        (40, 40, 40),
        2,
        cv2.LINE_AA
    )


# ============================================================
# 22. EDGE-AWARE COORDINATE MAPPING
# ============================================================

def map_finger_to_canvas(tip, width, height):

    usable_width = 1.0 - (2.0 * EDGE_MARGIN)
    usable_height = 1.0 - (2.0 * EDGE_MARGIN)

    normalized_x = (
        (tip.x - EDGE_MARGIN) / usable_width
    )

    normalized_y = (
        (tip.y - EDGE_MARGIN) / usable_height
    )

    # Keep floating-point coordinates for better accuracy.

    x = normalized_x * (width - 1)
    y = normalized_y * (height - 1)

    x = max(0.0, min(width - 1.0, x))
    y = max(0.0, min(height - 1.0, y))

    return x, y


# ============================================================
# 23. MAIN APPLICATION
# ============================================================

try:

    with mp_hands.Hands(

        static_image_mode=False,
        max_num_hands=1,
        model_complexity=1,
        min_detection_confidence=MIN_DETECTION_CONFIDENCE,
        min_tracking_confidence=MIN_TRACKING_CONFIDENCE

    ) as hands:

        while True:

            success, frame = cap.read()

            if not success:

                print("Unable to read camera frame.")
                break

            frame = cv2.flip(frame, 1)

            height, width = frame.shape[:2]

            if canvas is None:
                canvas = np.full_like(frame, 255)

            # HAND DETECTION

            small_frame = cv2.resize(
                frame,
                (PROCESS_WIDTH, PROCESS_HEIGHT),
                interpolation=cv2.INTER_AREA
            )

            rgb = cv2.cvtColor(
                small_frame,
                cv2.COLOR_BGR2RGB
            )

            rgb.flags.writeable = False

            results = hands.process(rgb)

            rgb.flags.writeable = True

            hand_detected = bool(
                results.multi_hand_landmarks
            )

            status = "NO HAND DETECTED - SHOW YOUR HAND"
            status_color = (0, 0, 255)

            cursor_point = None

            debug_frame = small_frame.copy()

            # HAND FOUND

            if hand_detected:

                hand = results.multi_hand_landmarks[0]

                mp_drawing.draw_landmarks(
                    debug_frame,
                    hand,
                    mp_hands.HAND_CONNECTIONS
                )

                tip = hand.landmark[8]

                x, y = map_finger_to_canvas(
                    tip,
                    width,
                    height
                )

                lost_hand_frames = 0

                raw_extended = is_index_finger_extended(hand)

                index_is_extended = update_gesture(
                    raw_extended
                )

                # BEFORE START

                if not drawing_started:

                    status = "HAND DETECTED - PRESS S TO START"
                    status_color = (0, 150, 0)

                # PAUSED

                elif drawing_paused:

                    status = "PAUSED - PRESS P TO RESUME"
                    status_color = (0, 0, 255)

                    save_current_stroke()
                    reset_tracking()

                # DRAWING MODE

                elif index_is_extended:

                    status = "DRAWING"
                    status_color = brush_color

                    stable_point, actual_movement, movement_started = (
                        stabilize_finger((x, y))
                    )

                    smoothed_point = stable_point

                    cursor_point = (
                        int(round(stable_point[0])),
                        int(round(stable_point[1]))
                    )

                    # START OR CONTINUE STROKE

                    if movement_started:

                        if not current_stroke:

                            current_stroke_color = brush_color
                            current_stroke_tool = selected_tool

                            start_point = (
                                float(stationary_anchor[0]),
                                float(stationary_anchor[1])
                            )

                            current_stroke.append(start_point)

                        last_point = current_stroke[-1]

                        previous_point = (
                            float(last_point[0]),
                            float(last_point[1])
                        )

                    # ADD DRAWING POINTS

                    if (
                        actual_movement
                        and previous_point is not None
                    ):

                        movement = point_distance(
                            stable_point,
                            previous_point
                        )

                        if movement >= MIN_POINT_DISTANCE:

                            add_drawing_point(
                                current_stroke,
                                stable_point
                            )

                            previous_point = stable_point

                # INDEX FINGER NOT EXTENDED

                else:

                    status = "HAND DETECTED - EXTEND INDEX FINGER"
                    status_color = (0, 140, 255)

                    save_current_stroke()

                    previous_point = None
                    smoothed_point = None

                    reset_movement_filter()

            # NO HAND DETECTED

            else:

                lost_hand_frames += 1

                if lost_hand_frames >= MAX_LOST_HAND_FRAMES:

                    save_current_stroke()
                    reset_tracking()

            # DEBUG WINDOW

            debug_text = (
                "HAND FOUND" if hand_detected else "NO HAND"
            )

            debug_color = (
                (0, 255, 0) if hand_detected else (0, 0, 255)
            )

            cv2.putText(
                debug_frame,
                debug_text,
                (10, 25),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                debug_color,
                2,
                cv2.LINE_AA
            )

            if SHOW_DEBUG_WINDOW:

                cv2.imshow(
                    "Hand Tracking Debug",
                    debug_frame
                )

            # CANVAS DISPLAY

            output = canvas.copy()

            if current_stroke:

                draw_smooth_stroke(
                    output,
                    current_stroke,
                    current_stroke_color,
                    current_stroke_tool
                )

            draw_color_palette(
                output,
                selected_color_index
            )

            draw_tool_status(output)

            # DRAWING CURSOR

            if (
                cursor_point is not None
                and drawing_started
                and not drawing_paused
            ):

                cv2.circle(
                    output,
                    cursor_point,
                    max(
                        3,
                        TOOLS[selected_tool]["thickness"] // 2
                    ),
                    brush_color,
                    -1,
                    cv2.LINE_AA
                )

            # STATUS

            cv2.putText(
                output,
                status,
                (20, 100),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                status_color,
                2,
                cv2.LINE_AA
            )

            instructions = (
                "S: Start | 1-8: Color | N: Pen | M: Marker | "
                "X: Max | L: Pencil | B: Brush | "
                "P: Pause | Z: Undo | C: Clear | Q: Quit"
            )

            cv2.putText(
                output,
                instructions,
                (20, height - 20),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.43,
                (80, 80, 80),
                1,
                cv2.LINE_AA
            )

            cv2.imshow(
                "Virtual Paint App",
                output
            )

            # KEYBOARD CONTROLS

            key = cv2.waitKey(1) & 0xFF

            if ord("1") <= key <= ord("8"):

                selected_color_index = key - ord("1")
                brush_color = COLORS[selected_color_index][1]

            elif key == ord("n"):

                selected_tool = "PEN"

            elif key == ord("m"):

                selected_tool = "MARKER"

            elif key == ord("x"):

                selected_tool = "MAX THICKNESS"

            elif key == ord("l"):

                selected_tool = "PENCIL"

            elif key == ord("b"):

                selected_tool = "BRUSH"

            elif key == ord("s") and not drawing_started:

                drawing_started = True
                drawing_paused = False

                reset_tracking()
                save_current_stroke()

            elif key == ord("p") and drawing_started:

                drawing_paused = not drawing_paused

                save_current_stroke()
                reset_tracking()

            elif key == ord("z") and drawing_started:

                save_current_stroke()

                if undo_history:

                    last_stroke = undo_history.pop()

                    for i in range(len(strokes) - 1, -1, -1):

                        if strokes[i] is last_stroke:

                            strokes.pop(i)
                            break

                    rebuild_canvas()

                reset_tracking()

            elif key == ord("c"):

                strokes.clear()
                current_stroke.clear()
                undo_history.clear()

                canvas[:] = 255

                drawing_started = False
                drawing_paused = False

                lost_hand_frames = 0

                reset_tracking()

            elif key == ord("q"):

                break

finally:

    cap.release()
    cv2.destroyAllWindows()
```

```python

import cv2
import mediapipe as mp
import numpy as np
import math
from collections import deque

# ============================================================
# 1. CAMERA SETUP
# ============================================================

CAMERA_INDEX = 0
FRAME_WIDTH = 1280
FRAME_HEIGHT = 720

cap = cv2.VideoCapture(CAMERA_INDEX)

if not cap.isOpened():
    cap.release()
    cap = cv2.VideoCapture(1)

if not cap.isOpened():
    raise RuntimeError("Camera not found. Check your camera connection.")

cap.set(cv2.CAP_PROP_FRAME_WIDTH, FRAME_WIDTH)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, FRAME_HEIGHT)
cap.set(cv2.CAP_PROP_FPS, 30)
cap.set(cv2.CAP_PROP_BUFFERSIZE, 1)

mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils

# ============================================================
# 2. COLORS
# ============================================================

COLORS = [
    ("Green", (0, 200, 0)),
    ("Red", (0, 0, 255)),
    ("Blue", (255, 0, 0)),
    ("Black", (0, 0, 0)),
    ("Yellow", (0, 255, 255)),
    ("Purple", (180, 0, 180)),
    ("Orange", (0, 140, 255)),
    ("Pink", (203, 192, 255))
]

selected_color_index = 0
brush_color = COLORS[selected_color_index][1]

# ============================================================
# 3. DRAWING TOOLS
# ============================================================

TOOLS = {
    "PEN": {"thickness": 2, "opacity": 1.0},
    "MARKER": {"thickness": 9, "opacity": 0.55},
    "MAX THICKNESS": {"thickness": 24, "opacity": 1.0},
    "PENCIL": {"thickness": 2, "opacity": 0.80},
    "BRUSH": {"thickness": 15, "opacity": 0.90}
}

selected_tool = "PEN"
current_stroke_tool = selected_tool
current_stroke_color = brush_color

# ============================================================
# 4. CANVAS VARIABLES
# ============================================================

canvas = None

strokes = []
current_stroke = []
undo_history = []

MAX_UNDO = 10

drawing_started = False
drawing_paused = False

previous_point = None
smoothed_point = None

# ============================================================
# 5. HAND DETECTION SETTINGS
# ============================================================

PROCESS_WIDTH = 640
PROCESS_HEIGHT = 360

MIN_DETECTION_CONFIDENCE = 0.60
MIN_TRACKING_CONFIDENCE = 0.65

UP_THRESHOLD = 2
DOWN_THRESHOLD = 3

finger_up_count = 0
finger_down_count = 0
index_is_extended = False

lost_hand_frames = 0
MAX_LOST_HAND_FRAMES = 2

ANGLE_THRESHOLD_2D = 125
ANGLE_THRESHOLD_3D = 130
LENGTH_RATIO_THRESHOLD = 1.00

SHOW_DEBUG_WINDOW = True

# ============================================================
# 6. DRAWING QUALITY SETTINGS
# ============================================================

# Responsive smoothing
MIN_SMOOTHING = 0.58
MAX_SMOOTHING = 0.84

# Capture small finger movements
MIN_POINT_DISTANCE = 0.65

# Smooth curve interpolation
CURVE_STEPS = 16

# Cursor
CURSOR_RADIUS = 6

# Movement detection
MOVEMENT_START_THRESHOLD = 2.0

# Noise reduction
RAW_HISTORY_SIZE = 3
NOISE_DEAD_ZONE = 0.18

# Drawing boundaries
EDGE_MARGIN = 0.035

raw_point_history = deque(maxlen=RAW_HISTORY_SIZE)

stationary_anchor = None
filtered_point = None
movement_active = False

# ============================================================
# 7. DISTANCE FUNCTIONS
# ============================================================

def distance_2d(a, b):
    return math.hypot(a.x - b.x, a.y - b.y)


def distance_3d(a, b):
    return math.sqrt(
        (a.x - b.x) ** 2 +
        (a.y - b.y) ** 2 +
        (a.z - b.z) ** 2
    )


def point_distance(p1, p2):
    return math.hypot(
        p1[0] - p2[0],
        p1[1] - p2[1]
    )


# ============================================================
# 8. ANGLE FUNCTIONS
# ============================================================

def calculate_angle_2d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


def calculate_angle_3d(a, b, c):

    v1 = np.array(
        [a.x - b.x, a.y - b.y, a.z - b.z],
        dtype=np.float32
    )

    v2 = np.array(
        [c.x - b.x, c.y - b.y, c.z - b.z],
        dtype=np.float32
    )

    denominator = np.linalg.norm(v1) * np.linalg.norm(v2)

    if denominator < 1e-8:
        return 0

    cosine = np.dot(v1, v2) / denominator
    cosine = np.clip(cosine, -1.0, 1.0)

    return math.degrees(math.acos(cosine))


# ============================================================
# 9. INDEX FINGER DETECTION
# ============================================================

def is_index_finger_extended(hand):

    lm = hand.landmark

    mcp = lm[5]
    pip = lm[6]
    dip = lm[7]
    tip = lm[8]

    angle_2d = calculate_angle_2d(mcp, pip, tip)
    angle_3d = calculate_angle_3d(mcp, pip, tip)

    dip_angle_2d = calculate_angle_2d(pip, dip, tip)
    dip_angle_3d = calculate_angle_3d(pip, dip, tip)

    ratio_2d = (
        distance_2d(mcp, tip) /
        max(distance_2d(mcp, pip), 0.0001)
    )

    ratio_3d = (
        distance_3d(mcp, tip) /
        max(distance_3d(mcp, pip), 0.0001)
    )

    angle_ok = (
        angle_2d >= ANGLE_THRESHOLD_2D or
        angle_3d >= ANGLE_THRESHOLD_3D
    )

    length_ok = (
        ratio_2d >= LENGTH_RATIO_THRESHOLD or
        ratio_3d >= LENGTH_RATIO_THRESHOLD
    )

    tip_ok = (
        dip_angle_2d >= 110 or
        dip_angle_3d >= 110
    )

    return angle_ok and length_ok and tip_ok


# ============================================================
# 10. GESTURE STABILIZATION
# ============================================================

def update_gesture(raw_extended):

    global finger_up_count
    global finger_down_count
    global index_is_extended

    if raw_extended:
        finger_up_count += 1
        finger_down_count = 0
    else:
        finger_down_count += 1
        finger_up_count = 0

    if finger_up_count >= UP_THRESHOLD:
        index_is_extended = True

    elif finger_down_count >= DOWN_THRESHOLD:
        index_is_extended = False

    return index_is_extended


# ============================================================
# 11. RESET FILTERS
# ============================================================

def reset_movement_filter():

    global stationary_anchor
    global filtered_point
    global movement_active

    raw_point_history.clear()

    stationary_anchor = None
    filtered_point = None
    movement_active = False


def reset_tracking():

    global previous_point
    global smoothed_point
    global finger_up_count
    global finger_down_count
    global index_is_extended

    previous_point = None
    smoothed_point = None

    finger_up_count = 0
    finger_down_count = 0

    index_is_extended = False

    reset_movement_filter()


# ============================================================
# 12. RAW POINT NOISE REDUCTION
# ============================================================

def get_median_point(raw_point):

    raw_point_history.append(raw_point)

    points = np.array(
        list(raw_point_history),
        dtype=np.float32
    )

    return (
        float(np.median(points[:, 0])),
        float(np.median(points[:, 1]))
    )


# ============================================================
# 13. ADAPTIVE SMOOTHING
# ============================================================

def adaptive_smooth(new_point, old_point):

    if old_point is None:
        return (
            float(new_point[0]),
            float(new_point[1])
        )

    dx = new_point[0] - old_point[0]
    dy = new_point[1] - old_point[1]

    movement = math.hypot(dx, dy)

    if movement < NOISE_DEAD_ZONE:
        return old_point

    # Faster movement receives a higher alpha.
    # This reduces lag and improves responsiveness.

    speed_factor = min(movement / 18.0, 1.0)

    alpha = (
        MIN_SMOOTHING +
        (MAX_SMOOTHING - MIN_SMOOTHING) * speed_factor
    )

    return (
        old_point[0] + alpha * dx,
        old_point[1] + alpha * dy
    )


# ============================================================
# 14. CONTINUOUS MOVEMENT FILTER
# ============================================================

def stabilize_finger(raw_point):

    global stationary_anchor
    global filtered_point
    global movement_active

    clean_point = get_median_point(raw_point)

    if stationary_anchor is None:

        stationary_anchor = clean_point
        filtered_point = clean_point

        return clean_point, False, False

    # Wait for intentional movement before starting.

    if not movement_active:

        movement = point_distance(
            clean_point,
            stationary_anchor
        )

        if movement < MOVEMENT_START_THRESHOLD:
            return filtered_point, False, False

        movement_active = True

        filtered_point = clean_point

        return filtered_point, True, True

    # Never stop a stroke just because the finger becomes still.
    # This prevents broken circles and unwanted jumps.

    filtered_point = adaptive_smooth(
        clean_point,
        filtered_point
    )

    return filtered_point, True, False


# ============================================================
# 15. ADD DRAWING POINTS
# ============================================================

def add_drawing_point(points, new_point):

    if not points:

        points.append((
            float(new_point[0]),
            float(new_point[1])
        ))

        return

    last_point = points[-1]

    distance = point_distance(last_point, new_point)

    if distance >= MIN_POINT_DISTANCE:

        points.append((
            float(new_point[0]),
            float(new_point[1])
        ))


# ============================================================
# 16. CENTRIPETAL CATMULL-ROM CURVE
# ============================================================

def catmull_rom(p0, p1, p2, p3):

    p0 = np.array(p0, dtype=np.float64)
    p1 = np.array(p1, dtype=np.float64)
    p2 = np.array(p2, dtype=np.float64)
    p3 = np.array(p3, dtype=np.float64)

    alpha = 0.5
    epsilon = 1e-4

    def get_t(t, a, b):

        distance = np.linalg.norm(b - a)

        return t + max(distance, epsilon) ** alpha

    t0 = 0.0
    t1 = get_t(t0, p0, p1)
    t2 = get_t(t1, p1, p2)
    t3 = get_t(t2, p2, p3)

    result = []

    for i in range(CURVE_STEPS):

        t = i / CURVE_STEPS
        current_t = t1 + t * (t2 - t1)

        A1 = (
            ((t1 - current_t) / (t1 - t0)) * p0
            +
            ((current_t - t0) / (t1 - t0)) * p1
        )

        A2 = (
            ((t2 - current_t) / (t2 - t1)) * p1
            +
            ((current_t - t1) / (t2 - t1)) * p2
        )

        A3 = (
            ((t3 - current_t) / (t3 - t2)) * p2
            +
            ((current_t - t2) / (t3 - t2)) * p3
        )

        B1 = (
            ((t2 - current_t) / (t2 - t0)) * A1
            +
            ((current_t - t0) / (t2 - t0)) * A2
        )

        B2 = (
            ((t3 - current_t) / (t3 - t1)) * A2
            +
            ((current_t - t1) / (t3 - t1)) * A3
        )

        point = (
            ((t2 - current_t) / (t2 - t1)) * B1
            +
            ((current_t - t1) / (t2 - t1)) * B2
        )

        # Do not clip the curve.
        # Clipping can flatten circles and natural curves.

        result.append((
            float(point[0]),
            float(point[1])
        ))

    return result


def get_smooth_curve(points):

    if len(points) <= 2:

        return [
            (int(round(p[0])), int(round(p[1])))
            for p in points
        ]

    result = []

    extended = (
        [points[0]] +
        points +
        [points[-1]]
    )

    for i in range(1, len(extended) - 2):

        curve_points = catmull_rom(
            extended[i - 1],
            extended[i],
            extended[i + 1],
            extended[i + 2]
        )

        if result:
            curve_points = curve_points[1:]

        result.extend(curve_points)

    result.append(points[-1])

    return [
        (int(round(p[0])), int(round(p[1])))
        for p in result
    ]


# ============================================================
# 17. HIGH QUALITY DRAWING RENDERING
# ============================================================

def draw_smooth_stroke(image, points, color, tool="PEN"):

    if not points:
        return

    settings = TOOLS.get(tool, TOOLS["PEN"])

    thickness = settings["thickness"]
    opacity = settings["opacity"]

    radius = max(1, thickness // 2)

    smooth_points = get_smooth_curve(points)

    # Separate layer and mask.
    # This also supports pure black color correctly.

    layer = np.zeros_like(image)
    mask = np.zeros(image.shape[:2], dtype=np.uint8)

    draw_color = color

    if tool == "PENCIL":

        draw_color = tuple(
            int(c * 0.85 + 255 * 0.15)
            for c in color
        )

    # Single point

    if len(smooth_points) == 1:

        cv2.circle(
            layer,
            smooth_points[0],
            radius,
            draw_color,
            -1,
            cv2.LINE_AA
        )

        cv2.circle(
            mask,
            smooth_points[0],
            radius,
            255,
            -1,
            cv2.LINE_AA
        )

    else:

        pts = np.array(
            smooth_points,
            dtype=np.int32
        ).reshape((-1, 1, 2))

        # Draw anti-aliased curve

        cv2.polylines(
            layer,
            [pts],
            False,
            draw_color,
            thickness,
            cv2.LINE_AA
        )

        cv2.polylines(
            mask,
            [pts],
            False,
            255,
            thickness,
            cv2.LINE_AA
        )

        # Rounded line ends

        for point in (
            smooth_points[0],
            smooth_points[-1]
        ):

            cv2.circle(
                layer,
                point,
                radius,
                draw_color,
                -1,
                cv2.LINE_AA
            )

            cv2.circle(
                mask,
                point,
                radius,
                255,
                -1,
                cv2.LINE_AA
            )

    # Soft brush effect

    if tool == "BRUSH":

        mask = cv2.GaussianBlur(
            mask,
            (5, 5),
            1.0
        )

    # Alpha blending

    alpha = (
        mask.astype(np.float32) / 255.0
    ) * opacity

    alpha = alpha[:, :, np.newaxis]

    blended = (
        layer.astype(np.float32) * alpha +
        image.astype(np.float32) * (1.0 - alpha)
    )

    image[:] = np.clip(
        blended,
        0,
        255
    ).astype(np.uint8)


# ============================================================
# 18. SAVE CURRENT STROKE
# ============================================================

def save_current_stroke():

    global current_stroke
    global strokes
    global undo_history

    if not current_stroke:
        return

    stroke = {
        "points": current_stroke.copy(),
        "color": current_stroke_color,
        "tool": current_stroke_tool
    }

    strokes.append(stroke)

    undo_history.append(stroke)
    undo_history = undo_history[-MAX_UNDO:]

    draw_smooth_stroke(
        canvas,
        stroke["points"],
        stroke["color"],
        stroke["tool"]
    )

    current_stroke = []


# ============================================================
# 19. REBUILD CANVAS
# ============================================================

def rebuild_canvas():

    canvas[:] = 255

    for stroke in strokes:

        draw_smooth_stroke(
            canvas,
            stroke["points"],
            stroke["color"],
            stroke["tool"]
        )


# ============================================================
# 20. COLOR PALETTE
# ============================================================

def draw_color_palette(image, selected_index):

    start_x = 20
    start_y = 23

    spacing = 32
    radius = 6

    cv2.rectangle(
        image,
        (8, 7),
        (start_x + 7 * spacing + 12, 48),
        (245, 245, 245),
        -1
    )

    for i, (name, color) in enumerate(COLORS):

        x = start_x + i * spacing
        y = start_y

        cv2.circle(
            image,
            (x, y),
            radius,
            color,
            -1,
            cv2.LINE_AA
        )

        if i == selected_index:

            cv2.circle(
                image,
                (x, y),
                radius + 3,
                (0, 0, 0),
                1,
                cv2.LINE_AA
            )

        cv2.putText(
            image,
            str(i + 1),
            (x - 3, y + 17),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.3,
            (0, 0, 0),
            1,
            cv2.LINE_AA
        )


# ============================================================
# 21. TOOL STATUS
# ============================================================

def draw_tool_status(image):

    text = (
        f"TOOL: {selected_tool} | "
        f"SIZE: {TOOLS[selected_tool]['thickness']}"
    )

    cv2.putText(
        image,
        text,
        (20, 135),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.55,
        (40, 40, 40),
        2,
        cv2.LINE_AA
    )


# ============================================================
# 22. COORDINATE MAPPING
# ============================================================

def map_finger_to_canvas(tip, width, height):

    usable_width = 1.0 - (2.0 * EDGE_MARGIN)
    usable_height = 1.0 - (2.0 * EDGE_MARGIN)

    normalized_x = (
        (tip.x - EDGE_MARGIN) / usable_width
    )

    normalized_y = (
        (tip.y - EDGE_MARGIN) / usable_height
    )

    x = normalized_x * (width - 1)
    y = normalized_y * (height - 1)

    x = max(0.0, min(width - 1.0, x))
    y = max(0.0, min(height - 1.0, y))

    return x, y


# ============================================================
# 23. MAIN APPLICATION
# ============================================================

try:

    with mp_hands.Hands(

        static_image_mode=False,
        max_num_hands=1,
        model_complexity=1,
        min_detection_confidence=MIN_DETECTION_CONFIDENCE,
        min_tracking_confidence=MIN_TRACKING_CONFIDENCE

    ) as hands:

        while True:

            success, frame = cap.read()

            if not success:
                print("Unable to read camera frame.")
                break

            frame = cv2.flip(frame, 1)

            height, width = frame.shape[:2]

            if canvas is None:
                canvas = np.full_like(frame, 255)

            # HAND DETECTION

            small_frame = cv2.resize(
                frame,
                (PROCESS_WIDTH, PROCESS_HEIGHT),
                interpolation=cv2.INTER_AREA
            )

            rgb = cv2.cvtColor(
                small_frame,
                cv2.COLOR_BGR2RGB
            )

            rgb.flags.writeable = False
            results = hands.process(rgb)
            rgb.flags.writeable = True

            hand_detected = bool(
                results.multi_hand_landmarks
            )

            status = "NO HAND DETECTED - SHOW YOUR HAND"
            status_color = (0, 0, 255)

            cursor_point = None

            debug_frame = small_frame.copy()

            # HAND FOUND

            if hand_detected:

                hand = results.multi_hand_landmarks[0]

                mp_drawing.draw_landmarks(
                    debug_frame,
                    hand,
                    mp_hands.HAND_CONNECTIONS
                )

                tip = hand.landmark[8]

                x, y = map_finger_to_canvas(
                    tip,
                    width,
                    height
                )

                lost_hand_frames = 0

                raw_extended = is_index_finger_extended(hand)

                index_is_extended = update_gesture(
                    raw_extended
                )

                # BEFORE START

                if not drawing_started:

                    status = "HAND DETECTED - PRESS S TO START"
                    status_color = (0, 150, 0)

                # PAUSED

                elif drawing_paused:

                    status = "PAUSED - PRESS P TO RESUME"
                    status_color = (0, 0, 255)

                    save_current_stroke()
                    reset_tracking()

                # DRAWING MODE

                elif index_is_extended:

                    status = "DRAWING"
                    status_color = brush_color

                    stable_point, actual_movement, movement_started = (
                        stabilize_finger((x, y))
                    )

                    smoothed_point = stable_point

                    cursor_point = (
                        int(round(stable_point[0])),
                        int(round(stable_point[1]))
                    )

                    # START NEW STROKE

                    if movement_started:

                        if not current_stroke:

                            current_stroke_color = brush_color
                            current_stroke_tool = selected_tool

                            start_point = (
                                float(stationary_anchor[0]),
                                float(stationary_anchor[1])
                            )

                            current_stroke.append(start_point)

                        last_point = current_stroke[-1]

                        previous_point = (
                            float(last_point[0]),
                            float(last_point[1])
                        )

                    # ADD POINTS TO STROKE

                    if (
                        actual_movement
                        and previous_point is not None
                    ):

                        movement = point_distance(
                            stable_point,
                            previous_point
                        )

                        if movement >= MIN_POINT_DISTANCE:

                            add_drawing_point(
                                current_stroke,
                                stable_point
                            )

                            previous_point = stable_point

                # INDEX FINGER NOT EXTENDED

                else:

                    status = "HAND DETECTED - EXTEND INDEX FINGER"
                    status_color = (0, 140, 255)

                    save_current_stroke()

                    previous_point = None
                    smoothed_point = None

                    reset_movement_filter()

            # NO HAND DETECTED

            else:

                lost_hand_frames += 1

                if lost_hand_frames >= MAX_LOST_HAND_FRAMES:

                    save_current_stroke()
                    reset_tracking()

            # DEBUG WINDOW

            debug_text = (
                "HAND FOUND" if hand_detected else "NO HAND"
            )

            debug_color = (
                (0, 255, 0) if hand_detected else (0, 0, 255)
            )

            cv2.putText(
                debug_frame,
                debug_text,
                (10, 25),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                debug_color,
                2,
                cv2.LINE_AA
            )

            if SHOW_DEBUG_WINDOW:

                cv2.imshow(
                    "Hand Tracking Debug",
                    debug_frame
                )

            # CANVAS DISPLAY

            output = canvas.copy()

            if current_stroke:

                draw_smooth_stroke(
                    output,
                    current_stroke,
                    current_stroke_color,
                    current_stroke_tool
                )

            draw_color_palette(
                output,
                selected_color_index
            )

            draw_tool_status(output)

            # DRAWING CURSOR

            if (
                cursor_point is not None
                and drawing_started
                and not drawing_paused
            ):

                cv2.circle(
                    output,
                    cursor_point,
                    max(
                        3,
                        TOOLS[selected_tool]["thickness"] // 2
                    ),
                    brush_color,
                    -1,
                    cv2.LINE_AA
                )

            # STATUS

            cv2.putText(
                output,
                status,
                (20, 100),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                status_color,
                2,
                cv2.LINE_AA
            )

            instructions = (
                "S: Start | 1-8: Color | N: Pen | M: Marker | "
                "X: Max | L: Pencil | B: Brush | "
                "P: Pause | Z: Undo | C: Clear | Q: Quit"
            )

            cv2.putText(
                output,
                instructions,
                (20, height - 20),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.43,
                (80, 80, 80),
                1,
                cv2.LINE_AA
            )

            cv2.imshow(
                "Virtual Paint App",
                output
            )

            # ====================================================
            # KEYBOARD CONTROLS
            # ====================================================

            key = cv2.waitKey(1) & 0xFF

            if ord("1") <= key <= ord("8"):

                selected_color_index = key - ord("1")
                brush_color = COLORS[selected_color_index][1]

            elif key == ord("n"):

                selected_tool = "PEN"

            elif key == ord("m"):

                selected_tool = "MARKER"

            elif key == ord("x"):

                selected_tool = "MAX THICKNESS"

            elif key == ord("l"):

                selected_tool = "PENCIL"

            elif key == ord("b"):

                selected_tool = "BRUSH"

            elif key == ord("s") and not drawing_started:

                drawing_started = True
                drawing_paused = False

                reset_tracking()
                save_current_stroke()

            elif key == ord("p") and drawing_started:

                drawing_paused = not drawing_paused

                save_current_stroke()
                reset_tracking()

            elif key == ord("z") and drawing_started:

                save_current_stroke()

                if undo_history:

                    last_stroke = undo_history.pop()

                    for i in range(len(strokes) - 1, -1, -1):

                        if strokes[i] is last_stroke:

                            strokes.pop(i)
                            break

                    rebuild_canvas()

                reset_tracking()

            elif key == ord("c"):

                strokes.clear()
                current_stroke.clear()
                undo_history.clear()

                canvas[:] = 255

                drawing_started = False
                drawing_paused = False

                lost_hand_frames = 0

                reset_tracking()

            elif key == ord("q"):

                break

finally:

    cap.release()
    cv2.destroyAllWindows()
```
